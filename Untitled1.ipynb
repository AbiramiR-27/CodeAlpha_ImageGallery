```python
# ============================================================
# CELL 3 — BUILD GROUND-TRUTH POSITIVE-PAIR INDEX
# ============================================================

import sqlite3
import pandas as pd
import os
import gc

DATASET_DIR = "/content/drive/MyDrive/dataset"
DB_PATH = (
    "/content/drive/MyDrive/"
    "amazon_training/databases/training_lookup.db"
)

# Remove an incomplete database if one exists
if os.path.exists(DB_PATH):
    os.remove(DB_PATH)

conn = sqlite3.connect(DB_PATH)

print("=" * 70)
print("BUILDING POSITIVE-PAIR INDEX")
print("=" * 70)

# ------------------------------------------------------------
# Create table
# ------------------------------------------------------------

conn.execute("""
CREATE TABLE positive_pairs (
    source1_entity_id TEXT NOT NULL,
    matched_entity_id TEXT NOT NULL,
    matched_source TEXT NOT NULL
)
""")

# ------------------------------------------------------------
# Read ground truth in chunks
# ------------------------------------------------------------

GT_FILE = f"{DATASET_DIR}/train_ground_truth.tsv"

total_pairs = 0

for chunk in pd.read_csv(
    GT_FILE,
    sep="\t",
    dtype=str,
    chunksize=100_000
):

    rows = []

    for row in chunk.itertuples(index=False):

        s1_id = row.source1_entity_id
        matches = str(row.matched_entity_ids)

        if not matches or matches == "nan":
            continue

        for candidate_id in matches.split(","):

            candidate_id = candidate_id.strip()

            if not candidate_id:
                continue

            if candidate_id.startswith("S2-"):
                source = "S2"

            elif candidate_id.startswith("S3-"):
                source = "S3"

            else:
                continue

            rows.append(
                (
                    s1_id,
                    candidate_id,
                    source
                )
            )

    if rows:

        conn.executemany(
            """
            INSERT INTO positive_pairs
            VALUES (?, ?, ?)
            """,
            rows
        )

        total_pairs += len(rows)

    del rows
    del chunk

    gc.collect()

    if total_pairs % 1_000_000 < 100_000:
        print(
            f"Positive pairs inserted: "
            f"{total_pairs:,}"
        )

conn.commit()

# ------------------------------------------------------------
# Indexes
# ------------------------------------------------------------

print("\nCreating indexes...")

conn.execute("""
CREATE INDEX idx_positive_s1
ON positive_pairs(source1_entity_id)
""")

conn.execute("""
CREATE INDEX idx_positive_candidate
ON positive_pairs(matched_entity_id)
""")

conn.execute("""
CREATE INDEX idx_positive_source
ON positive_pairs(matched_source)
""")

conn.commit()

# ------------------------------------------------------------
# Verify
# ------------------------------------------------------------

total = conn.execute("""
    SELECT COUNT(*)
    FROM positive_pairs
""").fetchone()[0]

s2 = conn.execute("""
    SELECT COUNT(*)
    FROM positive_pairs
    WHERE matched_source = 'S2'
""").fetchone()[0]

s3 = conn.execute("""
    SELECT COUNT(*)
    FROM positive_pairs
    WHERE matched_source = 'S3'
""").fetchone()[0]

s1 = conn.execute("""
    SELECT COUNT(DISTINCT source1_entity_id)
    FROM positive_pairs
""").fetchone()[0]

print("=" * 70)
print("POSITIVE-PAIR INDEX COMPLETE")
print("=" * 70)

print(f"Total positive pairs : {total:,}")
print(f"S2 positive pairs    : {s2:,}")
print(f"S3 positive pairs    : {s3:,}")
print(f"S1 with matches      : {s1:,}")
print(f"\nDatabase:")
print(DB_PATH)

conn.close()
```

```python
# ============================================================
# CELL 4 — BUILD SOURCE LOOKUP TABLES
# ============================================================

import sqlite3
import pandas as pd
import os
import gc

DATASET_DIR = "/content/drive/MyDrive/dataset"

DB_PATH = (
    "/content/drive/MyDrive/"
    "amazon_training/databases/training_lookup.db"
)

conn = sqlite3.connect(DB_PATH)

SOURCE_FILES = {
    "source1": "cleaned_train_source1.tsv",
    "source2": "cleaned_train_source2.tsv",
    "source3": "cleaned_train_source3.tsv"
}

COLUMNS = [
    "entity_id",
    "country_norm",
    "business_name_norm",
    "business_address_norm",
    "business_name_clean",
    "business_address_clean",
    "address_numbers"
]

for table_name, filename in SOURCE_FILES.items():

    print("=" * 70)
    print(f"BUILDING {table_name.upper()}")
    print("=" * 70)

    # Remove table if it already exists
    conn.execute(
        f"DROP TABLE IF EXISTS {table_name}"
    )

    conn.execute(f"""
        CREATE TABLE {table_name} (
            entity_id TEXT,
            country_norm TEXT,
            business_name_norm TEXT,
            business_address_norm TEXT,
            business_name_clean TEXT,
            business_address_clean TEXT,
            address_numbers TEXT
        )
    """)

    filepath = os.path.join(
        DATASET_DIR,
        filename
    )

    total_rows = 0

    for chunk in pd.read_csv(
        filepath,
        sep="\t",
        dtype=str,
        usecols=COLUMNS,
        chunksize=100_000,
        keep_default_na=False
    ):

        # Replace pandas missing values if any
        chunk = chunk.fillna("")

        rows = list(
            chunk.itertuples(
                index=False,
                name=None
            )
        )

        conn.executemany(
            f"""
            INSERT INTO {table_name}
            VALUES (?, ?, ?, ?, ?, ?, ?)
            """,
            rows
        )

        total_rows += len(rows)

        del rows
        del chunk

        gc.collect()

    conn.commit()

    print(
        f"Rows inserted: {total_rows:,}"
    )

    # Entity ID index
    print("Creating entity_id index...")

    conn.execute(f"""
        CREATE INDEX idx_{table_name}_id
        ON {table_name}(entity_id)
    """)

    conn.commit()

    print("Done.")

# ------------------------------------------------------------
# Verify
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("SOURCE TABLE VERIFICATION")
print("=" * 70)

for table_name in SOURCE_FILES:

    count = conn.execute(
        f"SELECT COUNT(*) FROM {table_name}"
    ).fetchone()[0]

    print(
        f"{table_name}: {count:,} rows"
    )

conn.close()

print("\nDatabase permanently stored at:")
print(DB_PATH)
```

```python
# ============================================================
# CELL 6 — GENERATE HARD NEGATIVE PAIRS
# ============================================================

import sqlite3
import pandas as pd
import numpy as np
import os
import random
import gc
from collections import defaultdict

DB_PATH = (
    "/content/drive/MyDrive/"
    "amazon_training/databases/training_lookup.db"
)

FEATURE_DIR = (
    "/content/drive/MyDrive/"
    "amazon_training/features"
)

random.seed(42)
np.random.seed(42)

N_NEG = 50_000
POOL_SIZE = 50

conn = sqlite3.connect(DB_PATH)

print("=" * 70)
print("LOADING POSITIVE PAIR KEYS")
print("=" * 70)

positive_keys = set(
    conn.execute("""
        SELECT
            source1_entity_id,
            matched_entity_id
        FROM positive_pairs
    """).fetchall()
)

print(
    f"Positive keys loaded: {len(positive_keys):,}"
)

# ------------------------------------------------------------
# STEP 1 — Load the sampled S1 entities
# ------------------------------------------------------------

positive_sample = pd.read_csv(
    os.path.join(
        FEATURE_DIR,
        "positive_pairs_50k.tsv"
    ),
    sep="\t",
    dtype=str
)

s1_ids = positive_sample[
    "source1_entity_id"
].drop_duplicates().tolist()

print(
    f"Sampled S1 entities: {len(s1_ids):,}"
)

# ------------------------------------------------------------
# STEP 2 — Get S1 name prefixes
# ------------------------------------------------------------

placeholders = ",".join(["?"] * len(s1_ids))

s1_df = pd.read_sql_query(
    f"""
    SELECT
        entity_id,
        business_name_norm
    FROM source1
    WHERE entity_id IN ({placeholders})
    """,
    conn,
    params=s1_ids
)

print(
    f"S1 records loaded: {len(s1_df):,}"
)

def get_prefix(x):
    x = str(x).strip().lower()
    return x[:3] if len(x) >= 3 else x

s1_df["prefix3"] = (
    s1_df["business_name_norm"]
    .map(get_prefix)
)

# ------------------------------------------------------------
# STEP 3 — Build bounded candidate pools
# ------------------------------------------------------------

print("\nBuilding bounded candidate pools...")

prefix_pools = {
    "S2": defaultdict(list),
    "S3": defaultdict(list)
}

for source, table in [
    ("S2", "source2"),
    ("S3", "source3")
]:

    print(f"Building {source} pools...")

    # Read only ID + normalized name
    for chunk in pd.read_sql_query(
        f"""
        SELECT
            entity_id,
            business_name_norm
        FROM {table}
        """,
        conn,
        chunksize=100_000
    ):

        chunk["prefix3"] = (
            chunk["business_name_norm"]
            .astype(str)
            .map(get_prefix)
        )

        for prefix, group in chunk.groupby("prefix3"):

            pool = prefix_pools[source][prefix]

            remaining = POOL_SIZE - len(pool)

            if remaining > 0:

                ids = group["entity_id"].tolist()

                if len(ids) > remaining:
                    ids = random.sample(
                        ids,
                        remaining
                    )

                pool.extend(ids)

        del chunk
        gc.collect()

    print(
        f"{source} pools: "
        f"{len(prefix_pools[source]):,}"
    )

# ------------------------------------------------------------
# STEP 4 — Generate negatives
# ------------------------------------------------------------

print("\nGenerating hard negatives...")

negative_rows = []

target_s2 = N_NEG // 2
target_s3 = N_NEG - target_s2

s2_count = 0
s3_count = 0

for _, row in s1_df.iterrows():

    s1_id = row["entity_id"]
    prefix = row["prefix3"]

    # -------------------------
    # S2 negative
    # -------------------------

    if s2_count < target_s2:

        candidates = prefix_pools["S2"].get(
            prefix,
            []
        )

        random.shuffle(candidates)

        for candidate_id in candidates:

            if (
                s1_id,
                candidate_id
            ) not in positive_keys:

                negative_rows.append(
                    (
                        s1_id,
                        candidate_id,
                        "S2"
                    )
                )

                s2_count += 1
                break

    # -------------------------
    # S3 negative
    # -------------------------

    if s3_count < target_s3:

        candidates = prefix_pools["S3"].get(
            prefix,
            []
        )

        random.shuffle(candidates)

        for candidate_id in candidates:

            if (
                s1_id,
                candidate_id
            ) not in positive_keys:

                negative_rows.append(
                    (
                        s1_id,
                        candidate_id,
                        "S3"
                    )
                )

                s3_count += 1
                break

    if (
        s2_count >= target_s2
        and s3_count >= target_s3
    ):
        break

# ------------------------------------------------------------
# STEP 5 — Save
# ------------------------------------------------------------

negative_pairs = pd.DataFrame(
    negative_rows,
    columns=[
        "source1_entity_id",
        "matched_entity_id",
        "matched_source"
    ]
)

print("\n" + "=" * 70)
print("NEGATIVE PAIR RESULTS")
print("=" * 70)

print(
    f"Negative pairs generated: "
    f"{len(negative_pairs):,}"
)

print("\nSource distribution:")
print(
    negative_pairs[
        "matched_source"
    ].value_counts()
)

negative_path = os.path.join(
    FEATURE_DIR,
    "negative_pairs_50k.tsv"
)

negative_pairs.to_csv(
    negative_path,
    sep="\t",
    index=False
)

print("\nSaved:")
print(negative_path)

conn.close()

del (
    positive_keys,
    positive_sample,
    s1_df,
    prefix_pools,
    negative_pairs
)

gc.collect()

print("\nCELL 6 COMPLETE")
```

```python
# ============================================================
# CELL 7 — BUILD TRAINING FEATURES
# ============================================================

import pandas as pd
import numpy as np
import sqlite3
import os
import gc
import re

from rapidfuzz.fuzz import (
    ratio,
    token_sort_ratio,
    token_set_ratio
)

DB_PATH = (
    "/content/drive/MyDrive/"
    "amazon_training/databases/training_lookup.db"
)

FEATURE_DIR = (
    "/content/drive/MyDrive/"
    "amazon_training/features"
)

POS_PATH = os.path.join(
    FEATURE_DIR,
    "positive_pairs_50k.tsv"
)

NEG_PATH = os.path.join(
    FEATURE_DIR,
    "negative_pairs_50k.tsv"
)

OUTPUT_PATH = os.path.join(
    FEATURE_DIR,
    "training_features_100k.tsv"
)

# ------------------------------------------------------------
# Load pairs
# ------------------------------------------------------------

print("=" * 70)
print("LOADING TRAINING PAIRS")
print("=" * 70)

positive = pd.read_csv(
    POS_PATH,
    sep="\t",
    dtype=str
)

negative = pd.read_csv(
    NEG_PATH,
    sep="\t",
    dtype=str
)

positive["label"] = 1
negative["label"] = 0

pairs = pd.concat(
    [positive, negative],
    ignore_index=True
)

print(
    f"Total pairs: {len(pairs):,}"
)

print(
    f"Positive: {(pairs['label'] == 1).sum():,}"
)

print(
    f"Negative: {(pairs['label'] == 0).sum():,}"
)

del positive, negative
gc.collect()

# ------------------------------------------------------------
# Connect to database
# ------------------------------------------------------------

conn = sqlite3.connect(DB_PATH)

# ------------------------------------------------------------
# Feature functions
# ------------------------------------------------------------

def number_set(text):
    return set(
        re.findall(
            r"\d+",
            str(text)
        )
    )

def number_overlap(a, b):

    a_nums = number_set(a)
    b_nums = number_set(b)

    if not a_nums and not b_nums:
        return 1.0

    if not a_nums or not b_nums:
        return 0.0

    return len(
        a_nums & b_nums
    ) / len(
        a_nums | b_nums
    )

def calculate_features(row):

    name1 = str(row["s1_business_name_norm"])
    name2 = str(row["match_business_name_norm"])

    clean_name1 = str(row["s1_business_name_clean"])
    clean_name2 = str(row["match_business_name_clean"])

    addr1 = str(row["s1_business_address_norm"])
    addr2 = str(row["match_business_address_norm"])

    clean_addr1 = str(row["s1_business_address_clean"])
    clean_addr2 = str(row["match_business_address_clean"])

    return {
        "name_exact":
            int(name1 == name2),

        "name_clean_exact":
            int(clean_name1 == clean_name2),

        "name_ratio":
            ratio(name1, name2) / 100.0,

        "name_token_sort":
            token_sort_ratio(name1, name2) / 100.0,

        "name_token_set":
            token_set_ratio(name1, name2) / 100.0,

        "address_exact":
            int(addr1 == addr2),

        "address_clean_exact":
            int(clean_addr1 == clean_addr2),

        "address_ratio":
            ratio(addr1, addr2) / 100.0,

        "address_token_sort":
            token_sort_ratio(addr1, addr2) / 100.0,

        "address_token_set":
            token_set_ratio(addr1, addr2) / 100.0,

        "number_overlap":
            number_overlap(addr1, addr2),

        "name_length_diff":
            abs(len(name1) - len(name2)),

        "address_length_diff":
            abs(len(addr1) - len(addr2)),

        "country_match":
            int(
                row["s1_country_norm"]
                ==
                row["match_country_norm"]
            ),

        "source_type":
            2 if row["matched_source"] == "S2"
            else 3,

        "label":
            int(row["label"])
    }

# ------------------------------------------------------------
# Process in chunks
# ------------------------------------------------------------

CHUNK_SIZE = 10_000

all_features = []

print("\nComputing features...")

for start in range(
    0,
    len(pairs),
    CHUNK_SIZE
):

    chunk = pairs.iloc[
        start:start + CHUNK_SIZE
    ].copy()

    # -------------------------------
    # Get S1 records
    # -------------------------------

    s1_ids = chunk[
        "source1_entity_id"
    ].tolist()

    placeholders = ",".join(
        ["?"] * len(s1_ids)
    )

    s1_df = pd.read_sql_query(
        f"""
        SELECT
            entity_id,
            country_norm,
            business_name_norm,
            business_address_norm,
            business_name_clean,
            business_address_clean
        FROM source1
        WHERE entity_id IN ({placeholders})
        """,
        conn,
        params=s1_ids
    )

    s1_df = s1_df.rename(
        columns={
            "entity_id":
                "source1_entity_id",

            "country_norm":
                "s1_country_norm",

            "business_name_norm":
                "s1_business_name_norm",

            "business_address_norm":
                "s1_business_address_norm",

            "business_name_clean":
                "s1_business_name_clean",

            "business_address_clean":
                "s1_business_address_clean"
        }
    )

    # -------------------------------
    # Get S2/S3 records
    # -------------------------------

    enriched = []

    for source, group in chunk.groupby(
        "matched_source"
    ):

        table = (
            "source2"
            if source == "S2"
            else "source3"
        )

        ids = group[
            "matched_entity_id"
        ].tolist()

        placeholders = ",".join(
            ["?"] * len(ids)
        )

        match_df = pd.read_sql_query(
            f"""
            SELECT
                entity_id,
                country_norm,
                business_name_norm,
                business_address_norm,
                business_name_clean,
                business_address_clean
            FROM {table}
            WHERE entity_id IN ({placeholders})
            """,
            conn,
            params=ids
        )

        match_df = match_df.rename(
            columns={
                "entity_id":
                    "matched_entity_id",

                "country_norm":
                    "match_country_norm",

                "business_name_norm":
                    "match_business_name_norm",

                "business_address_norm":
                    "match_business_address_norm",

                "business_name_clean":
                    "match_business_name_clean",

                "business_address_clean":
                    "match_business_address_clean"
            }
        )

        match_df["matched_source"] = source

        enriched.append(match_df)

    match_all = pd.concat(
        enriched,
        ignore_index=True
    )

    # -------------------------------
    # Merge
    # -------------------------------

    chunk = chunk.merge(
        s1_df,
        on="source1_entity_id",
        how="left"
    )

    chunk = chunk.merge(
        match_all,
        on=[
            "matched_entity_id",
            "matched_source"
        ],
        how="left"
    )

    # -------------------------------
    # Calculate features
    # -------------------------------

    feature_rows = []

    for _, row in chunk.iterrows():

        feature_rows.append(
            calculate_features(row)
        )

    feature_df = pd.DataFrame(
        feature_rows
    )

    all_features.append(
        feature_df
    )

    print(
        f"Processed "
        f"{min(start + CHUNK_SIZE, len(pairs)):,}"
        f"/{len(pairs):,}"
    )

    del (
        chunk,
        s1_df,
        match_all,
        enriched,
        feature_df,
        feature_rows
    )

    gc.collect()

# ------------------------------------------------------------
# Combine
# ------------------------------------------------------------

features = pd.concat(
    all_features,
    ignore_index=True
)

print("\n" + "=" * 70)
print("FEATURE GENERATION COMPLETE")
print("=" * 70)

print(
    f"Feature rows: {len(features):,}"
)

print(
    f"Feature columns: {features.shape[1]}"
)

print("\nLabel distribution:")
print(
    features["label"].value_counts()
)

# ------------------------------------------------------------
# Save
# ------------------------------------------------------------

features.to_csv(
    OUTPUT_PATH,
    sep="\t",
    index=False
)

print("\nSaved permanently to:")
print(OUTPUT_PATH)

conn.close()

del pairs, features, all_features
gc.collect()

print("\nCELL 7 COMPLETE")
```

```python
# ============================================================
# CELL 8 — RESTORE PAIR IDs + GROUPED TRAIN/VALID SPLIT
# ============================================================

import pandas as pd
import numpy as np
import os
import gc

FEATURE_DIR = (
    "/content/drive/MyDrive/"
    "amazon_training/features"
)

FEATURE_PATH = os.path.join(
    FEATURE_DIR,
    "training_features_100k.tsv"
)

POS_PATH = os.path.join(
    FEATURE_DIR,
    "positive_pairs_50k.tsv"
)

NEG_PATH = os.path.join(
    FEATURE_DIR,
    "negative_pairs_50k.tsv"
)

TRAIN_PATH = os.path.join(
    FEATURE_DIR,
    "train_features.tsv"
)

VALID_PATH = os.path.join(
    FEATURE_DIR,
    "valid_features.tsv"
)

# ------------------------------------------------------------
# 1. Load saved features
# ------------------------------------------------------------

print("=" * 70)
print("LOADING SAVED FEATURES")
print("=" * 70)

features = pd.read_csv(
    FEATURE_PATH,
    sep="\t"
)

print(
    f"Feature rows: {len(features):,}"
)

# ------------------------------------------------------------
# 2. Load pair identifiers
# ------------------------------------------------------------

positive_pairs = pd.read_csv(
    POS_PATH,
    sep="\t",
    dtype=str
)

negative_pairs = pd.read_csv(
    NEG_PATH,
    sep="\t",
    dtype=str
)

# Keep only the identifiers needed for grouping
positive_ids = positive_pairs[
    [
        "source1_entity_id",
        "matched_entity_id",
        "matched_source"
    ]
].copy()

negative_ids = negative_pairs[
    [
        "source1_entity_id",
        "matched_entity_id",
        "matched_source"
    ]
].copy()

# Same order as Cell 7:
# positives first, negatives second
pair_ids = pd.concat(
    [positive_ids, negative_ids],
    ignore_index=True
)

print(
    f"Pair ID rows: {len(pair_ids):,}"
)

# ------------------------------------------------------------
# 3. Safety checks
# ------------------------------------------------------------

assert len(features) == len(pair_ids), (
    f"Row mismatch: "
    f"{len(features)} features vs "
    f"{len(pair_ids)} pair IDs"
)

print("Row alignment check: PASSED")

# ------------------------------------------------------------
# 4. Attach identifiers
# ------------------------------------------------------------

features.insert(
    0,
    "source1_entity_id",
    pair_ids["source1_entity_id"].values
)

features.insert(
    1,
    "matched_entity_id",
    pair_ids["matched_entity_id"].values
)

features.insert(
    2,
    "matched_source",
    pair_ids["matched_source"].values
)

print(
    f"Combined dataset columns: "
    f"{features.shape[1]}"
)

# ------------------------------------------------------------
# 5. Grouped 80/20 split by S1 entity
# ------------------------------------------------------------

rng = np.random.default_rng(42)

unique_s1 = features[
    "source1_entity_id"
].unique()

rng.shuffle(unique_s1)

split_point = int(
    len(unique_s1) * 0.80
)

train_s1 = set(
    unique_s1[:split_point]
)

valid_s1 = set(
    unique_s1[split_point:]
)

train_df = features[
    features["source1_entity_id"].isin(train_s1)
].copy()

valid_df = features[
    features["source1_entity_id"].isin(valid_s1)
].copy()

# ------------------------------------------------------------
# 6. Report
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("GROUPED SPLIT RESULTS")
print("=" * 70)

print(
    f"Total S1 entities: {len(unique_s1):,}"
)

print(
    f"Train S1 entities: {len(train_s1):,}"
)

print(
    f"Validation S1 entities: {len(valid_s1):,}"
)

print(
    f"Train pairs: {len(train_df):,}"
)

print(
    f"Validation pairs: {len(valid_df):,}"
)

print("\nTrain labels:")
print(
    train_df["label"].value_counts()
)

print("\nValidation labels:")
print(
    valid_df["label"].value_counts()
)

# ------------------------------------------------------------
# 7. Leakage check
# ------------------------------------------------------------

overlap = (
    set(train_df["source1_entity_id"])
    &
    set(valid_df["source1_entity_id"])
)

print(
    "\nS1 entities appearing in both sets:",
    len(overlap)
)

assert len(overlap) == 0

# ------------------------------------------------------------
# 8. Save
# ------------------------------------------------------------

train_df.to_csv(
    TRAIN_PATH,
    sep="\t",
    index=False
)

valid_df.to_csv(
    VALID_PATH,
    sep="\t",
    index=False
)

print("\nSaved:")
print(TRAIN_PATH)
print(VALID_PATH)

del (
    features,
    positive_pairs,
    negative_pairs,
    positive_ids,
    negative_ids,
    pair_ids,
    train_df,
    valid_df
)

gc.collect()

print("\nCELL 8 COMPLETE")
```

```python
# ============================================================
# CELL 9 — TRAIN LIGHTGBM MODEL
# ============================================================

import pandas as pd
import numpy as np
import lightgbm as lgb
import os
import gc

FEATURE_DIR = (
    "/content/drive/MyDrive/"
    "amazon_training/features"
)

MODEL_DIR = (
    "/content/drive/MyDrive/"
    "amazon_training/models"
)

os.makedirs(MODEL_DIR, exist_ok=True)

TRAIN_PATH = os.path.join(
    FEATURE_DIR,
    "train_features.tsv"
)

VALID_PATH = os.path.join(
    FEATURE_DIR,
    "valid_features.tsv"
)

MODEL_PATH = os.path.join(
    MODEL_DIR,
    "entity_match_lightgbm.txt"
)

# ------------------------------------------------------------
# Load
# ------------------------------------------------------------

print("=" * 70)
print("LOADING TRAIN / VALIDATION DATA")
print("=" * 70)

train_df = pd.read_csv(
    TRAIN_PATH,
    sep="\t"
)

valid_df = pd.read_csv(
    VALID_PATH,
    sep="\t"
)

FEATURES = [
    "name_exact",
    "name_clean_exact",
    "name_ratio",
    "name_token_sort",
    "name_token_set",
    "address_exact",
    "address_clean_exact",
    "address_ratio",
    "address_token_sort",
    "address_token_set",
    "number_overlap",
    "name_length_diff",
    "address_length_diff",
    "country_match",
    "source_type"
]

X_train = train_df[FEATURES]
y_train = train_df["label"]

X_valid = valid_df[FEATURES]
y_valid = valid_df["label"]

print(f"Train rows: {len(X_train):,}")
print(f"Valid rows: {len(X_valid):,}")
print(f"Features: {len(FEATURES)}")

# ------------------------------------------------------------
# LightGBM datasets
# ------------------------------------------------------------

train_data = lgb.Dataset(
    X_train,
    label=y_train,
    feature_name=FEATURES
)

valid_data = lgb.Dataset(
    X_valid,
    label=y_valid,
    feature_name=FEATURES,
    reference=train_data
)

# ------------------------------------------------------------
# Model parameters
# ------------------------------------------------------------

params = {
    "objective": "binary",
    "metric": "binary_logloss",
    "boosting_type": "gbdt",

    "learning_rate": 0.05,

    "num_leaves": 31,
    "max_depth": -1,

    "min_child_samples": 30,

    "subsample": 0.8,
    "subsample_freq": 1,

    "colsample_bytree": 0.9,

    "reg_alpha": 0.1,
    "reg_lambda": 1.0,

    "verbosity": -1,
    "seed": 42
}

print("\n" + "=" * 70)
print("TRAINING LIGHTGBM")
print("=" * 70)

model = lgb.train(
    params,
    train_data,
    num_boost_round=1000,

    valid_sets=[
        train_data,
        valid_data
    ],

    valid_names=[
        "train",
        "valid"
    ],

    callbacks=[
        lgb.early_stopping(
            stopping_rounds=50
        ),
        lgb.log_evaluation(
            period=50
        )
    ]
)

# ------------------------------------------------------------
# Save model immediately
# ------------------------------------------------------------

model.save_model(
    MODEL_PATH
)

print("\n" + "=" * 70)
print("MODEL SAVED")
print("=" * 70)

print(
    "Best iteration:",
    model.best_iteration
)

print(
    "Model path:",
    MODEL_PATH
)

# ------------------------------------------------------------
# Validation probabilities
# ------------------------------------------------------------

valid_pred = model.predict(
    X_valid,
    num_iteration=model.best_iteration
)

print(
    "\nValidation predictions generated:",
    len(valid_pred)
)

del (
    train_df,
    valid_df,
    X_train,
    X_valid,
    y_train,
    y_valid,
    train_data,
    valid_data
)

gc.collect()

print("\nCELL 9 COMPLETE")
```

```python
# ============================================================
# CELL 10 — VALIDATION F0.5 + THRESHOLD OPTIMIZATION
# ============================================================

import os
import json
import pandas as pd
import numpy as np
import lightgbm as lgb

from sklearn.metrics import (
    precision_score,
    recall_score,
    fbeta_score
)

MODEL_PATH = (
    "/content/drive/MyDrive/"
    "amazon_training/models/"
    "entity_match_lightgbm.txt"
)

VALID_PATH = (
    "/content/drive/MyDrive/"
    "amazon_training/features/"
    "valid_features.tsv"
)

THRESHOLD_PATH = (
    "/content/drive/MyDrive/"
    "amazon_training/models/"
    "best_threshold.json"
)

FEATURES = [
    "name_exact",
    "name_clean_exact",
    "name_ratio",
    "name_token_sort",
    "name_token_set",
    "address_exact",
    "address_clean_exact",
    "address_ratio",
    "address_token_sort",
    "address_token_set",
    "number_overlap",
    "name_length_diff",
    "address_length_diff",
    "country_match",
    "source_type"
]

# ------------------------------------------------------------
# 1. Verify model
# ------------------------------------------------------------

print("=" * 70)
print("VERIFYING MODEL")
print("=" * 70)

if not os.path.exists(MODEL_PATH):
    raise FileNotFoundError(
        f"Model not found:\n{MODEL_PATH}"
    )

print("Model exists:")
print(MODEL_PATH)

model = lgb.Booster(
    model_file=MODEL_PATH
)

print(
    "Number of trees:",
    model.num_trees()
)

# ------------------------------------------------------------
# 2. Load validation data
# ------------------------------------------------------------

valid_df = pd.read_csv(
    VALID_PATH,
    sep="\t"
)

X_valid = valid_df[FEATURES]
y_valid = valid_df["label"]

print("\nValidation rows:")
print(len(valid_df))

# ------------------------------------------------------------
# 3. Predict
# ------------------------------------------------------------

valid_prob = model.predict(
    X_valid
)

print(
    "Predictions generated:",
    len(valid_prob)
)

# ------------------------------------------------------------
# 4. Search threshold
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("SEARCHING F0.5 THRESHOLD")
print("=" * 70)

results = []

thresholds = np.arange(
    0.05,
    0.951,
    0.01
)

for threshold in thresholds:

    pred = (
        valid_prob >= threshold
    ).astype(int)

    precision = precision_score(
        y_valid,
        pred,
        zero_division=0
    )

    recall = recall_score(
        y_valid,
        pred,
        zero_division=0
    )

    f05 = fbeta_score(
        y_valid,
        pred,
        beta=0.5,
        zero_division=0
    )

    results.append({
        "threshold": threshold,
        "precision": precision,
        "recall": recall,
        "f0.5": f05,
        "predicted_positive": int(
            pred.sum()
        )
    })

results_df = pd.DataFrame(results)

best_row = results_df.loc[
    results_df["f0.5"].idxmax()
]

print("\nBEST THRESHOLD")
print("-" * 70)

print(
    f"Threshold: "
    f"{best_row['threshold']:.2f}"
)

print(
    f"Precision: "
    f"{best_row['precision']:.6f}"
)

print(
    f"Recall: "
    f"{best_row['recall']:.6f}"
)

print(
    f"F0.5: "
    f"{best_row['f0.5']:.6f}"
)

print(
    f"Predicted positives: "
    f"{int(best_row['predicted_positive']):,}"
)

# ------------------------------------------------------------
# 5. Show top thresholds
# ------------------------------------------------------------

print("\nTOP 10 THRESHOLDS")
print("-" * 70)

print(
    results_df
    .sort_values(
        "f0.5",
        ascending=False
    )
    .head(10)
    .to_string(index=False)
)

# ------------------------------------------------------------
# 6. Save threshold
# ------------------------------------------------------------

best_threshold = float(
    best_row["threshold"]
)

with open(
    THRESHOLD_PATH,
    "w"
) as f:

    json.dump(
        {
            "threshold": best_threshold,
            "metric": "F0.5",
            "validation_rows": len(valid_df)
        },
        f,
        indent=2
    )

print("\nSaved threshold:")
print(THRESHOLD_PATH)

print("\nCELL 10 COMPLETE")
```

```python
# ============================================================
# CELL 11 — BOUNDED TEST CANDIDATE GENERATION
# ============================================================

import os
import gc
import random
import pandas as pd
from collections import defaultdict

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

TEST_DIR = (
    "/content/drive/MyDrive/"
    "amazon_training/test_prepared"
)

CANDIDATE_DIR = (
    "/content/drive/MyDrive/"
    "amazon_training/candidates"
)

os.makedirs(
    CANDIDATE_DIR,
    exist_ok=True
)

OUTPUT_PATH = os.path.join(
    CANDIDATE_DIR,
    "candidate_pairs.tsv"
)

# ------------------------------------------------------------
# TEST FILES
# ------------------------------------------------------------

TEST_FILES = {
    "S1": "test_source1_prepared.tsv",
    "S2": "test_source2_prepared.tsv",
    "S3": "test_source3_prepared.tsv"
}

for source, filename in TEST_FILES.items():

    path = os.path.join(
        TEST_DIR,
        filename
    )

    print(
        f"{source}: {path}"
    )

    if not os.path.exists(path):
        raise FileNotFoundError(
            f"\nMissing {source} file:\n{path}\n\n"
            f"Check the filename inside {TEST_DIR}"
        )

print("\nAll prepared test files found.")

# ------------------------------------------------------------
# PARAMETERS
# ------------------------------------------------------------

# Maximum candidates retained for each
# S1 prefix/source combination.
POOL_SIZE = 50

CHUNK_SIZE = 100_000

random.seed(42)

# ------------------------------------------------------------
# HELPER
# ------------------------------------------------------------

def prefix3(value):

    value = str(value).strip().lower()

    if len(value) < 3:
        return value

    return value[:3]


# ------------------------------------------------------------
# STEP 1 — BUILD S2/S3 PREFIX POOLS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("BUILDING BOUNDED CANDIDATE POOLS")
print("=" * 70)

prefix_pools = {
    "S2": defaultdict(list),
    "S3": defaultdict(list)
}

for source in ["S2", "S3"]:

    filepath = os.path.join(
        TEST_DIR,
        TEST_FILES[source]
    )

    print(
        f"\nProcessing {source}:"
    )

    rows_seen = 0

    for chunk in pd.read_csv(
        filepath,
        sep="\t",
        dtype=str,
        usecols=[
            "entity_id",
            "business_name_norm"
        ],
        chunksize=CHUNK_SIZE,
        keep_default_na=False
    ):

        chunk = chunk.fillna("")

        chunk["prefix3"] = (
            chunk["business_name_norm"]
            .map(prefix3)
        )

        for prefix, group in chunk.groupby(
            "prefix3"
        ):

            pool = prefix_pools[source][
                prefix
            ]

            remaining = (
                POOL_SIZE - len(pool)
            )

            if remaining <= 0:
                continue

            ids = group[
                "entity_id"
            ].tolist()

            if len(ids) > remaining:

                ids = random.sample(
                    ids,
                    remaining
                )

            pool.extend(ids)

        rows_seen += len(chunk)

        del chunk
        gc.collect()

    print(
        f"Rows scanned: {rows_seen:,}"
    )

    print(
        f"Unique prefixes: "
        f"{len(prefix_pools[source]):,}"
    )

# ------------------------------------------------------------
# STEP 2 — GENERATE CANDIDATE PAIRS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("GENERATING CANDIDATE PAIRS")
print("=" * 70)

# Remove old file if it exists.
if os.path.exists(OUTPUT_PATH):

    os.remove(
        OUTPUT_PATH
    )

header_written = False

total_candidates = 0
s1_processed = 0

s1_path = os.path.join(
    TEST_DIR,
    TEST_FILES["S1"]
)

for chunk in pd.read_csv(
    s1_path,
    sep="\t",
    dtype=str,
    usecols=[
        "entity_id",
        "business_name_norm"
    ],
    chunksize=CHUNK_SIZE,
    keep_default_na=False
):

    chunk = chunk.fillna("")

    candidate_rows = []

    for row in chunk.itertuples(
        index=False
    ):

        s1_id = row.entity_id

        p = prefix3(
            row.business_name_norm
        )

        # -------------------------
        # S2 candidates
        # -------------------------

        for candidate_id in (
            prefix_pools["S2"].get(
                p,
                []
            )
        ):

            candidate_rows.append(
                (
                    s1_id,
                    candidate_id,
                    "S2"
                )
            )

        # -------------------------
        # S3 candidates
        # -------------------------

        for candidate_id in (
            prefix_pools["S3"].get(
                p,
                []
            )
        ):

            candidate_rows.append(
                (
                    s1_id,
                    candidate_id,
                    "S3"
                )
            )

    candidate_df = pd.DataFrame(
        candidate_rows,
        columns=[
            "source1_entity_id",
            "matched_entity_id",
            "matched_source"
        ]
    )

    # --------------------------------------------------------
    # Remove duplicates within this chunk
    # --------------------------------------------------------

    candidate_df.drop_duplicates(
        inplace=True
    )

    # --------------------------------------------------------
    # Append to Drive
    # --------------------------------------------------------

    candidate_df.to_csv(
        OUTPUT_PATH,
        sep="\t",
        index=False,
        mode="a",
        header=not header_written
    )

    header_written = True

    total_candidates += len(
        candidate_df
    )

    s1_processed += len(chunk)

    print(
        f"S1 processed: "
        f"{s1_processed:,} | "
        f"Candidates written: "
        f"{total_candidates:,}"
    )

    del (
        chunk,
        candidate_rows,
        candidate_df
    )

    gc.collect()

# ------------------------------------------------------------
# FINAL VERIFICATION
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("CANDIDATE GENERATION COMPLETE")
print("=" * 70)

print(
    f"Total candidate rows: "
    f"{total_candidates:,}"
)

print(
    f"Candidate file: "
    f"{OUTPUT_PATH}"
)

print(
    f"File size: "
    f"{os.path.getsize(OUTPUT_PATH) / (1024**3):.2f} GB"
)

# Count source distribution
candidate_counts = pd.read_csv(
    OUTPUT_PATH,
    sep="\t",
    usecols=["matched_source"]
)["matched_source"].value_counts()

print("\nCandidate source distribution:")
print(candidate_counts)

print("\nCELL 11 COMPLETE")
```

```python
# ============================================================
# CELL 11A — PREPARE TEST FILES PERMANENTLY
# ============================================================

import os
import re
import gc
import pandas as pd
import unicodedata

TEST_RAW_DIR = (
    "/content/drive/MyDrive/test"
)

TEST_PREPARED_DIR = (
    "/content/drive/MyDrive/"
    "amazon_training/test_prepared"
)

os.makedirs(
    TEST_PREPARED_DIR,
    exist_ok=True
)

TEST_FILES = {
    "S1": "test_source1.tsv",
    "S2": "test_source2.tsv",
    "S3": "test_source3.tsv"
}

CHUNK_SIZE = 100_000


# ------------------------------------------------------------
# TEXT NORMALIZATION
# ------------------------------------------------------------

def normalize_text(text):

    text = str(text)

    text = unicodedata.normalize(
        "NFKC",
        text
    )

    text = text.lower()

    text = re.sub(
        r"[^\w\s]",
        " ",
        text,
        flags=re.UNICODE
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    )

    return text.strip()


def extract_numbers(text):

    return " ".join(
        re.findall(
            r"\d+",
            str(text)
        )
    )


# ------------------------------------------------------------
# PROCESS EACH FILE
# ------------------------------------------------------------

for source, filename in TEST_FILES.items():

    input_path = os.path.join(
        TEST_RAW_DIR,
        filename
    )

    output_path = os.path.join(
        TEST_PREPARED_DIR,
        f"test_{source.lower()}_prepared.tsv"
    )

    print("\n" + "=" * 70)
    print(f"PREPARING {source}")
    print("=" * 70)

    print("Input:")
    print(input_path)

    print("Output:")
    print(output_path)

    if not os.path.exists(input_path):
        raise FileNotFoundError(
            f"Missing input file:\n{input_path}"
        )

    # Remove old incomplete output if present
    if os.path.exists(output_path):
        os.remove(output_path)

    total_rows = 0
    header_written = False

    for chunk in pd.read_csv(
        input_path,
        sep="\t",
        dtype=str,
        chunksize=CHUNK_SIZE,
        keep_default_na=False
    ):

        chunk = chunk.fillna("")

        # -----------------------------
        # Normalize fields
        # -----------------------------

        chunk["country_norm"] = (
            chunk["country"]
            .map(normalize_text)
        )

        chunk["business_name_norm"] = (
            chunk["business_name"]
            .map(normalize_text)
        )

        chunk["business_address_norm"] = (
            chunk["business_address"]
            .map(normalize_text)
        )

        chunk["business_name_clean"] = (
            chunk["business_name_norm"]
            .str.replace(
                r"\s+",
                "",
                regex=True
            )
        )

        chunk["business_address_clean"] = (
            chunk["business_address_norm"]
            .str.replace(
                r"\s+",
                "",
                regex=True
            )
        )

        chunk["address_numbers"] = (
            chunk["business_address_norm"]
            .map(extract_numbers)
        )

        # -----------------------------
        # Keep only required columns
        # -----------------------------

        output_columns = [
            "entity_id",
            "country_norm",
            "business_name_norm",
            "business_address_norm",
            "business_name_clean",
            "business_address_clean",
            "address_numbers"
        ]

        chunk = chunk[
            output_columns
        ]

        # -----------------------------
        # Save chunk
        # -----------------------------

        chunk.to_csv(
            output_path,
            sep="\t",
            index=False,
            mode="a",
            header=not header_written
        )

        header_written = True

        total_rows += len(chunk)

        print(
            f"Processed: "
            f"{total_rows:,} rows"
        )

        del chunk
        gc.collect()

    print(
        f"\n{source} COMPLETE"
    )

    print(
        f"Total rows: {total_rows:,}"
    )

    print(
        f"Output size: "
        f"{os.path.getsize(output_path) / (1024**3):.2f} GB"
    )


# ------------------------------------------------------------
# FINAL CHECK
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("TEST PREPARATION COMPLETE")
print("=" * 70)

for source in ["S1", "S2", "S3"]:

    path = os.path.join(
        TEST_PREPARED_DIR,
        f"test_{source.lower()}_prepared.tsv"
    )

    print(
        f"{source}: {path}"
    )

print("\nCELL 11A COMPLETE")
```

```python
# ============================================================
# CELL 11B — BOUNDED TEST CANDIDATE GENERATION
# ============================================================

import os
import gc
import random
import pandas as pd
from collections import defaultdict

TEST_DIR = (
    "/content/drive/MyDrive/"
    "amazon_training/test_prepared"
)

CANDIDATE_DIR = (
    "/content/drive/MyDrive/"
    "amazon_training/candidates"
)

os.makedirs(
    CANDIDATE_DIR,
    exist_ok=True
)

OUTPUT_PATH = os.path.join(
    CANDIDATE_DIR,
    "candidate_pairs.tsv"
)

random.seed(42)

POOL_SIZE = 50
CHUNK_SIZE = 100_000


def prefix3(value):
    value = str(value).strip().lower()
    return value[:3]


# ============================================================
# STEP 1 — BUILD S2/S3 BOUNDED POOLS
# ============================================================

print("=" * 70)
print("BUILDING BOUNDED CANDIDATE POOLS")
print("=" * 70)

prefix_pools = {
    "S2": defaultdict(list),
    "S3": defaultdict(list)
}

for source in ["S2", "S3"]:

    filepath = os.path.join(
        TEST_DIR,
        f"test_{source.lower()}_prepared.tsv"
    )

    print(f"\nBuilding {source} pool...")

    rows_seen = 0

    for chunk in pd.read_csv(
        filepath,
        sep="\t",
        dtype=str,
        usecols=[
            "entity_id",
            "business_name_norm"
        ],
        chunksize=CHUNK_SIZE,
        keep_default_na=False
    ):

        chunk = chunk.fillna("")

        chunk["prefix3"] = (
            chunk["business_name_norm"]
            .map(prefix3)
        )

        for prefix, group in chunk.groupby(
            "prefix3"
        ):

            pool = prefix_pools[source][prefix]

            remaining = POOL_SIZE - len(pool)

            if remaining <= 0:
                continue

            ids = group["entity_id"].tolist()

            if len(ids) > remaining:
                ids = random.sample(
                    ids,
                    remaining
                )

            pool.extend(ids)

        rows_seen += len(chunk)

        del chunk
        gc.collect()

    print(
        f"Rows scanned: {rows_seen:,}"
    )

    print(
        f"Unique prefixes: "
        f"{len(prefix_pools[source]):,}"
    )


# ============================================================
# STEP 2 — GENERATE CANDIDATES
# ============================================================

print("\n" + "=" * 70)
print("GENERATING CANDIDATE_PAIRS.TSV")
print("=" * 70)

if os.path.exists(OUTPUT_PATH):
    os.remove(OUTPUT_PATH)

header_written = False

total_candidates = 0
s1_processed = 0

s1_path = os.path.join(
    TEST_DIR,
    "test_s1_prepared.tsv"
)

for chunk in pd.read_csv(
    s1_path,
    sep="\t",
    dtype=str,
    usecols=[
        "entity_id",
        "business_name_norm"
    ],
    chunksize=CHUNK_SIZE,
    keep_default_na=False
):

    chunk = chunk.fillna("")

    candidate_rows = []

    for row in chunk.itertuples(
        index=False
    ):

        s1_id = row.entity_id
        prefix = prefix3(
            row.business_name_norm
        )

        # S2 candidates
        for candidate_id in prefix_pools[
            "S2"
        ].get(prefix, []):

            candidate_rows.append(
                (
                    s1_id,
                    candidate_id,
                    "S2"
                )
            )

        # S3 candidates
        for candidate_id in prefix_pools[
            "S3"
        ].get(prefix, []):

            candidate_rows.append(
                (
                    s1_id,
                    candidate_id,
                    "S3"
                )
            )

    candidate_df = pd.DataFrame(
        candidate_rows,
        columns=[
            "source1_entity_id",
            "matched_entity_id",
            "matched_source"
        ]
    )

    candidate_df.drop_duplicates(
        inplace=True
    )

    candidate_df.to_csv(
        OUTPUT_PATH,
        sep="\t",
        index=False,
        mode="a",
        header=not header_written
    )

    header_written = True

    total_candidates += len(candidate_df)
    s1_processed += len(chunk)

    print(
        f"S1 processed: "
        f"{s1_processed:,} | "
        f"Candidates: "
        f"{total_candidates:,}"
    )

    del (
        chunk,
        candidate_rows,
        candidate_df
    )

    gc.collect()


# ============================================================
# STEP 3 — VERIFY
# ============================================================

print("\n" + "=" * 70)
print("CANDIDATE GENERATION COMPLETE")
print("=" * 70)

file_size_gb = (
    os.path.getsize(OUTPUT_PATH)
    / (1024 ** 3)
)

print(
    f"Total candidate rows: "
    f"{total_candidates:,}"
)

print(
    f"Candidate file size: "
    f"{file_size_gb:.2f} GB"
)

print(
    f"Saved to:\n{OUTPUT_PATH}"
)

print("\nCandidate source distribution:")

source_counts = pd.read_csv(
    OUTPUT_PATH,
    sep="\t",
    usecols=["matched_source"]
)["matched_source"].value_counts()

print(source_counts)

print("\nCELL 11B COMPLETE")
```

```python
# ============================================================
# CELL 14 — BUILD MULTI-KEY BLOCKING INDEXES
# ============================================================

import sqlite3
import pandas as pd
import re
import gc
from collections import defaultdict

DB_PATH = (
    "/content/drive/MyDrive/"
    "amazon_training/databases/"
    "training_lookup.db"
)

conn = sqlite3.connect(DB_PATH)

print("=" * 70)
print("BUILDING MULTI-KEY BLOCKING INDEXES")
print("=" * 70)


# ------------------------------------------------------------
# KEY FUNCTIONS
# ------------------------------------------------------------

def prefix3(x):
    x = str(x).strip()
    return x[:3]


def clean_prefix4(x):
    x = str(x).strip()
    return x[:4]


def first_number(x):
    nums = re.findall(r"\d+", str(x))
    return nums[0] if nums else ""


# ------------------------------------------------------------
# INDEX STRUCTURE
# ------------------------------------------------------------

indexes = {
    "S2": {
        "name_exact": defaultdict(set),
        "name_prefix3": defaultdict(set),
        "clean_prefix4": defaultdict(set),
        "number": defaultdict(set)
    },
    "S3": {
        "name_exact": defaultdict(set),
        "name_prefix3": defaultdict(set),
        "clean_prefix4": defaultdict(set),
        "number": defaultdict(set)
    }
}


# ------------------------------------------------------------
# BUILD INDEXES
# ------------------------------------------------------------

for source in ["S2", "S3"]:

    table = (
        "source2"
        if source == "S2"
        else "source3"
    )

    print(f"\nBuilding {source} indexes...")

    rows_seen = 0

    for chunk in pd.read_sql_query(
        f"""
        SELECT
            entity_id,
            business_name_norm,
            business_name_clean,
            address_numbers
        FROM {table}
        """,
        conn,
        chunksize=200_000
    ):

        chunk = chunk.fillna("")

        for row in chunk.itertuples(
            index=False
        ):

            entity_id = row.entity_id

            name = str(
                row.business_name_norm
            )

            clean_name = str(
                row.business_name_clean
            )

            numbers = str(
                row.address_numbers
            )

            # Exact normalized name
            if name:
                indexes[source][
                    "name_exact"
                ][name].add(entity_id)

                # First 3 normalized characters
                indexes[source][
                    "name_prefix3"
                ][prefix3(name)].add(entity_id)

            # First 4 cleaned characters
            if clean_name:
                indexes[source][
                    "clean_prefix4"
                ][clean_prefix4(clean_name)].add(
                    entity_id
                )

            # First address number
            number = first_number(numbers)

            if number:
                indexes[source][
                    "number"
                ][number].add(entity_id)

        rows_seen += len(chunk)

        del chunk
        gc.collect()

    print(
        f"Rows indexed: {rows_seen:,}"
    )

    for key in indexes[source]:

        print(
            f"  {key}: "
            f"{len(indexes[source][key]):,} keys"
        )


# ------------------------------------------------------------
# SAVE INDEXES IN MEMORY FOR NEXT CELL
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("INDEX BUILD COMPLETE")
print("=" * 70)

print(
    "Indexes are ready for recall evaluation."
)

conn.close()

print("\nCELL 14 COMPLETE")
```

```python
# ============================================================
# CELL 15 — MULTI-KEY BLOCKING RECALL
# ============================================================

import sqlite3
import pandas as pd
import re
import gc

DB_PATH = (
    "/content/drive/MyDrive/"
    "amazon_training/databases/"
    "training_lookup.db"
)

conn = sqlite3.connect(DB_PATH)

print("=" * 70)
print("MEASURING MULTI-KEY BLOCKING RECALL")
print("=" * 70)


# ------------------------------------------------------------
# Helper
# ------------------------------------------------------------

def first_number(x):
    nums = re.findall(r"\d+", str(x))
    return nums[0] if nums else ""


# ------------------------------------------------------------
# Check ground truth in chunks
# ------------------------------------------------------------

total_positive = 0

found_exact = 0
found_prefix3 = 0
found_clean4 = 0
found_number = 0
found_union = 0

# Track overlap of keys
found_exact_and_prefix = 0

print("\nChecking 7.6M ground-truth pairs...")

for chunk in pd.read_sql_query(
    """
    SELECT
        source1_entity_id,
        matched_entity_id,
        matched_source
    FROM positive_pairs
    """,
    conn,
    chunksize=100_000
):

    # --------------------------------------------------------
    # Load S1 information
    # --------------------------------------------------------

    s1_ids = chunk[
        "source1_entity_id"
    ].unique().tolist()

    placeholders = ",".join(
        ["?"] * len(s1_ids)
    )

    s1_df = pd.read_sql_query(
        f"""
        SELECT
            entity_id,
            business_name_norm,
            business_name_clean,
            address_numbers
        FROM source1
        WHERE entity_id IN ({placeholders})
        """,
        conn,
        params=s1_ids
    )

    s1_map = (
        s1_df
        .set_index("entity_id")
        .to_dict("index")
    )

    # --------------------------------------------------------
    # Load matching S2/S3 information
    # --------------------------------------------------------

    for source, group in chunk.groupby(
        "matched_source"
    ):

        table = (
            "source2"
            if source == "S2"
            else "source3"
        )

        match_ids = group[
            "matched_entity_id"
        ].unique().tolist()

        placeholders2 = ",".join(
            ["?"] * len(match_ids)
        )

        match_df = pd.read_sql_query(
            f"""
            SELECT
                entity_id,
                business_name_norm,
                business_name_clean,
                address_numbers
            FROM {table}
            WHERE entity_id IN ({placeholders2})
            """,
            conn,
            params=match_ids
        )

        match_map = (
            match_df
            .set_index("entity_id")
            .to_dict("index")
        )

        # ----------------------------------------------------
        # Check each positive
        # ----------------------------------------------------

        for row in group.itertuples(
            index=False
        ):

            total_positive += 1

            s1 = s1_map[
                row.source1_entity_id
            ]

            match = match_map[
                row.matched_entity_id
            ]

            name1 = str(
                s1["business_name_norm"]
            )

            name2 = str(
                match["business_name_norm"]
            )

            clean1 = str(
                s1["business_name_clean"]
            )

            clean2 = str(
                match["business_name_clean"]
            )

            nums1 = first_number(
                s1["address_numbers"]
            )

            nums2 = first_number(
                match["address_numbers"]
            )

            # -----------------------------
            # Key 1: exact normalized name
            # -----------------------------

            exact = (
                bool(name1)
                and name1 == name2
            )

            # -----------------------------
            # Key 2: prefix 3
            # -----------------------------

            prefix = (
                bool(name1)
                and
                name1[:3] == name2[:3]
            )

            # -----------------------------
            # Key 3: clean prefix 4
            # -----------------------------

            clean4 = (
                bool(clean1)
                and
                clean1[:4] == clean2[:4]
            )

            # -----------------------------
            # Key 4: address number
            # -----------------------------

            number = (
                bool(nums1)
                and
                nums1 == nums2
            )

            if exact:
                found_exact += 1

            if prefix:
                found_prefix3 += 1

            if clean4:
                found_clean4 += 1

            if number:
                found_number += 1

            if (
                exact
                or prefix
                or clean4
                or number
            ):
                found_union += 1

            if exact and prefix:
                found_exact_and_prefix += 1

        del match_df
        del match_map

    del chunk
    del s1_df
    del s1_map

    gc.collect()

    if total_positive % 1_000_000 < 100_000:

        print(
            f"Checked: "
            f"{total_positive:,}"
        )


# ------------------------------------------------------------
# RESULTS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("MULTI-KEY BLOCKING RECALL")
print("=" * 70)

print(
    f"Total positive pairs: "
    f"{total_positive:,}"
)

print()

print(
    f"Exact-name recall: "
    f"{found_exact / total_positive:.4%}"
)

print(
    f"Prefix-3 recall: "
    f"{found_prefix3 / total_positive:.4%}"
)

print(
    f"Clean-prefix-4 recall: "
    f"{found_clean4 / total_positive:.4%}"
)

print(
    f"Address-number recall: "
    f"{found_number / total_positive:.4%}"
)

print(
    f"\nUNION RECALL: "
    f"{found_union / total_positive:.4%}"
)

print(
    f"Pairs captured by union: "
    f"{found_union:,}"
)

print(
    f"Pairs still missed: "
    f"{total_positive - found_union:,}"
)

conn.close()

print("\nCELL 15 COMPLETE")
```

```python
# ============================================================
# CELL 15 — FAST MULTI-KEY BLOCKING RECALL
# ============================================================

import sqlite3
import pandas as pd
import re
import gc
import time

DB_PATH = (
    "/content/drive/MyDrive/"
    "amazon_training/databases/"
    "training_lookup.db"
)

conn = sqlite3.connect(DB_PATH)

print("=" * 70)
print("FAST MULTI-KEY BLOCKING RECALL")
print("=" * 70)

# ------------------------------------------------------------
# Helper
# ------------------------------------------------------------

def first_number(x):
    nums = re.findall(r"\d+", str(x))
    return nums[0] if nums else ""


# ------------------------------------------------------------
# Load positive pairs once
# ------------------------------------------------------------

print("\nLoading ground-truth positive pairs...")

gt = pd.read_sql_query("""
    SELECT
        source1_entity_id,
        matched_entity_id,
        matched_source
    FROM positive_pairs
""", conn)

print(f"Loaded: {len(gt):,} positive pairs")


# ------------------------------------------------------------
# Load source1 information once
# ------------------------------------------------------------

print("\nLoading Source 1...")

s1 = pd.read_sql_query("""
    SELECT
        entity_id,
        business_name_norm,
        business_name_clean,
        address_numbers
    FROM source1
""", conn)

s1 = s1.set_index("entity_id")

print(f"Source 1 loaded: {len(s1):,}")


# ------------------------------------------------------------
# Load S2/S3 information once
# ------------------------------------------------------------

print("\nLoading Source 2...")

s2 = pd.read_sql_query("""
    SELECT
        entity_id,
        business_name_norm,
        business_name_clean,
        address_numbers
    FROM source2
""", conn)

s2 = s2.set_index("entity_id")

print(f"Source 2 loaded: {len(s2):,}")


print("\nLoading Source 3...")

s3 = pd.read_sql_query("""
    SELECT
        entity_id,
        business_name_norm,
        business_name_clean,
        address_numbers
    FROM source3
""", conn)

s3 = s3.set_index("entity_id")

print(f"Source 3 loaded: {len(s3):,}")


# ------------------------------------------------------------
# Create feature columns directly
# ------------------------------------------------------------

print("\nCreating blocking keys...")

for df in [s1, s2, s3]:

    df["name_prefix3"] = (
        df["business_name_norm"]
        .fillna("")
        .astype(str)
        .str[:3]
    )

    df["clean_prefix4"] = (
        df["business_name_clean"]
        .fillna("")
        .astype(str)
        .str[:4]
    )

    df["first_number"] = (
        df["address_numbers"]
        .fillna("")
        .astype(str)
        .str.extract(r"(\d+)", expand=False)
        .fillna("")
    )


# ------------------------------------------------------------
# Merge ground truth with actual records
# ------------------------------------------------------------

print("\nJoining Source 1 information...")

gt = gt.merge(
    s1[
        [
            "business_name_norm",
            "business_name_clean",
            "name_prefix3",
            "clean_prefix4",
            "first_number"
        ]
    ],
    left_on="source1_entity_id",
    right_index=True,
    how="left"
)

gt.rename(columns={
    "business_name_norm": "s1_name",
    "business_name_clean": "s1_clean",
    "name_prefix3": "s1_prefix3",
    "clean_prefix4": "s1_clean4",
    "first_number": "s1_number"
}, inplace=True)


# ------------------------------------------------------------
# Split S2 / S3
# ------------------------------------------------------------

print("\nPreparing S2 positives...")

g2 = gt[gt["matched_source"] == "S2"].copy()

g2 = g2.merge(
    s2[
        [
            "business_name_norm",
            "business_name_clean",
            "name_prefix3",
            "clean_prefix4",
            "first_number"
        ]
    ],
    left_on="matched_entity_id",
    right_index=True,
    how="left"
)

g2.rename(columns={
    "business_name_norm": "m_name",
    "business_name_clean": "m_clean",
    "name_prefix3": "m_prefix3",
    "clean_prefix4": "m_clean4",
    "first_number": "m_number"
}, inplace=True)


print("Preparing S3 positives...")

g3 = gt[gt["matched_source"] == "S3"].copy()

g3 = g3.merge(
    s3[
        [
            "business_name_norm",
            "business_name_clean",
            "name_prefix3",
            "clean_prefix4",
            "first_number"
        ]
    ],
    left_on="matched_entity_id",
    right_index=True,
    how="left"
)

g3.rename(columns={
    "business_name_norm": "m_name",
    "business_name_clean": "m_clean",
    "name_prefix3": "m_prefix3",
    "clean_prefix4": "m_clean4",
    "first_number": "m_number"
}, inplace=True)


# ------------------------------------------------------------
# Calculate blocking recall
# ------------------------------------------------------------

def calculate_recall(df, label):

    total = len(df)

    exact = (
        (df["s1_name"] != "") &
        (df["s1_name"] == df["m_name"])
    )

    prefix3 = (
        (df["s1_prefix3"] != "") &
        (df["s1_prefix3"] == df["m_prefix3"])
    )

    clean4 = (
        (df["s1_clean4"] != "") &
        (df["s1_clean4"] == df["m_clean4"])
    )

    number = (
        (df["s1_number"] != "") &
        (df["s1_number"] == df["m_number"])
    )

    union = exact | prefix3 | clean4 | number

    print("\n" + "-" * 70)
    print(label)
    print("-" * 70)

    print(f"Total positives:       {total:,}")
    print(f"Exact-name recall:     {exact.mean():.4%}")
    print(f"Prefix-3 recall:       {prefix3.mean():.4%}")
    print(f"Clean-prefix-4 recall: {clean4.mean():.4%}")
    print(f"Address-number recall: {number.mean():.4%}")
    print()
    print(f"UNION RECALL:          {union.mean():.4%}")
    print(f"Pairs captured:        {union.sum():,}")
    print(f"Pairs missed:          {(~union).sum():,}")

    return union.sum(), total


start = time.time()

found2, total2 = calculate_recall(g2, "SOURCE 2")
found3, total3 = calculate_recall(g3, "SOURCE 3")

total_found = found2 + found3
total_positive = total2 + total3

print("\n" + "=" * 70)
print("FINAL MULTI-KEY BLOCKING RECALL")
print("=" * 70)

print(f"Total positive pairs: {total_positive:,}")
print(f"Pairs captured:       {total_found:,}")
print(f"Pairs missed:         {total_positive - total_found:,}")
print(f"UNION RECALL:         {total_found / total_positive:.4%}")

print(f"\nElapsed time: {(time.time() - start) / 60:.2f} minutes")

conn.close()

print("\nCELL 15 COMPLETE")
```

```python
# ============================================================
# CELL 15 — LOW-RAM SQLITE MULTI-KEY BLOCKING RECALL
# ============================================================

import sqlite3
import re
import time
import os

DB_PATH = (
    "/content/drive/MyDrive/"
    "amazon_training/databases/"
    "training_lookup.db"
)

print("=" * 70)
print("LOW-RAM MULTI-KEY BLOCKING RECALL")
print("=" * 70)

print("\nDatabase:", DB_PATH)
print("Size:", round(os.path.getsize(DB_PATH) / (1024**3), 2), "GB")

conn = sqlite3.connect(DB_PATH)

# ------------------------------------------------------------
# Make SQLite use disk for temporary operations
# ------------------------------------------------------------

conn.execute("PRAGMA temp_store = FILE")
conn.execute("PRAGMA cache_size = -50000")
conn.execute("PRAGMA synchronous = OFF")

# ------------------------------------------------------------
# SQLite function: first number from address_numbers
# ------------------------------------------------------------

def first_number(value):
    if value is None:
        return ""
    nums = re.findall(r"\d+", str(value))
    return nums[0] if nums else ""

conn.create_function(
    "first_number",
    1,
    first_number,
    deterministic=True
)

print("\nSQLite configured for low-RAM processing.")


# ============================================================
# FUNCTION TO MEASURE ONE SOURCE
# ============================================================

def measure_source(source_table, source_name):

    print("\n" + "=" * 70)
    print(f"MEASURING {source_name}")
    print("=" * 70)

    start = time.time()

    # --------------------------------------------------------
    # Count positives
    # --------------------------------------------------------

    total = conn.execute(
        f"""
        SELECT COUNT(*)
        FROM positive_pairs p
        WHERE p.matched_source = ?
        """,
        (source_name,)
    ).fetchone()[0]

    print(f"Positive pairs: {total:,}")

    # --------------------------------------------------------
    # Exact normalized name
    # --------------------------------------------------------

    print("Checking exact-name blocking...")

    exact = conn.execute(
        f"""
        SELECT COUNT(*)
        FROM positive_pairs p
        JOIN source1 a
            ON a.entity_id = p.source1_entity_id
        JOIN {source_table} b
            ON b.entity_id = p.matched_entity_id
        WHERE p.matched_source = ?
          AND a.business_name_norm != ''
          AND a.business_name_norm = b.business_name_norm
        """,
        (source_name,)
    ).fetchone()[0]

    print(f"Exact: {exact:,} ({exact / total:.4%})")

    # --------------------------------------------------------
    # Name prefix 3
    # --------------------------------------------------------

    print("Checking name-prefix-3 blocking...")

    prefix3 = conn.execute(
        f"""
        SELECT COUNT(*)
        FROM positive_pairs p
        JOIN source1 a
            ON a.entity_id = p.source1_entity_id
        JOIN {source_table} b
            ON b.entity_id = p.matched_entity_id
        WHERE p.matched_source = ?
          AND a.business_name_norm != ''
          AND substr(a.business_name_norm, 1, 3)
              = substr(b.business_name_norm, 1, 3)
        """,
        (source_name,)
    ).fetchone()[0]

    print(f"Prefix-3: {prefix3:,} ({prefix3 / total:.4%})")

    # --------------------------------------------------------
    # Clean name prefix 4
    # --------------------------------------------------------

    print("Checking clean-name-prefix-4 blocking...")

    clean4 = conn.execute(
        f"""
        SELECT COUNT(*)
        FROM positive_pairs p
        JOIN source1 a
            ON a.entity_id = p.source1_entity_id
        JOIN {source_table} b
            ON b.entity_id = p.matched_entity_id
        WHERE p.matched_source = ?
          AND a.business_name_clean != ''
          AND substr(a.business_name_clean, 1, 4)
              = substr(b.business_name_clean, 1, 4)
        """,
        (source_name,)
    ).fetchone()[0]

    print(f"Clean-prefix-4: {clean4:,} ({clean4 / total:.4%})")

    # --------------------------------------------------------
    # Address first-number blocking
    # --------------------------------------------------------

    print("Checking address-number blocking...")

    number = conn.execute(
        f"""
        SELECT COUNT(*)
        FROM positive_pairs p
        JOIN source1 a
            ON a.entity_id = p.source1_entity_id
        JOIN {source_table} b
            ON b.entity_id = p.matched_entity_id
        WHERE p.matched_source = ?
          AND first_number(a.address_numbers) != ''
          AND first_number(a.address_numbers)
              = first_number(b.address_numbers)
        """,
        (source_name,)
    ).fetchone()[0]

    print(f"Address-number: {number:,} ({number / total:.4%})")

    # --------------------------------------------------------
    # UNION RECALL
    # --------------------------------------------------------

    print("Checking UNION blocking recall...")

    union = conn.execute(
        f"""
        SELECT COUNT(*)
        FROM positive_pairs p
        JOIN source1 a
            ON a.entity_id = p.source1_entity_id
        JOIN {source_table} b
            ON b.entity_id = p.matched_entity_id
        WHERE p.matched_source = ?

          AND
          (
              (
                  a.business_name_norm != ''
                  AND a.business_name_norm = b.business_name_norm
              )

              OR

              (
                  a.business_name_norm != ''
                  AND substr(a.business_name_norm, 1, 3)
                      = substr(b.business_name_norm, 1, 3)
              )

              OR

              (
                  a.business_name_clean != ''
                  AND substr(a.business_name_clean, 1, 4)
                      = substr(b.business_name_clean, 1, 4)
              )

              OR

              (
                  first_number(a.address_numbers) != ''
                  AND first_number(a.address_numbers)
                      = first_number(b.address_numbers)
              )
          )
        """,
        (source_name,)
    ).fetchone()[0]

    missed = total - union

    print(f"UNION: {union:,} ({union / total:.4%})")
    print(f"Missed: {missed:,}")

    elapsed = (time.time() - start) / 60

    print(f"\nTime for {source_name}: {elapsed:.2f} minutes")

    return {
        "total": total,
        "exact": exact,
        "prefix3": prefix3,
        "clean4": clean4,
        "number": number,
        "union": union
    }


# ============================================================
# S2
# ============================================================

result_s2 = measure_source(
    "source2",
    "S2"
)


# ============================================================
# S3
# ============================================================

result_s3 = measure_source(
    "source3",
    "S3"
)


# ============================================================
# FINAL RESULT
# ============================================================

total = result_s2["total"] + result_s3["total"]

exact = result_s2["exact"] + result_s3["exact"]
prefix3 = result_s2["prefix3"] + result_s3["prefix3"]
clean4 = result_s2["clean4"] + result_s3["clean4"]
number = result_s2["number"] + result_s3["number"]
union = result_s2["union"] + result_s3["union"]

print("\n" + "=" * 70)
print("FINAL MULTI-KEY BLOCKING RECALL")
print("=" * 70)

print(f"Total positive pairs:   {total:,}")
print(f"Exact-name recall:      {exact / total:.4%}")
print(f"Prefix-3 recall:        {prefix3 / total:.4%}")
print(f"Clean-prefix-4 recall:  {clean4 / total:.4%}")
print(f"Address-number recall:  {number / total:.4%}")

print()
print(f"UNION RECALL:           {union / total:.4%}")
print(f"Pairs captured:         {union:,}")
print(f"Pairs missed:           {total - union:,}")

print("\n" + "=" * 70)
print("CELL 15 COMPLETE")
print("=" * 70)

conn.close()
```

```python
# ============================================================
# CELL 16 — ANALYZE MISSED POSITIVE PAIRS
# ============================================================

import sqlite3
import re
import time

DB_PATH = (
    "/content/drive/MyDrive/"
    "amazon_training/databases/"
    "training_lookup.db"
)

conn = sqlite3.connect(DB_PATH)

conn.execute("PRAGMA temp_store = FILE")
conn.execute("PRAGMA cache_size = -50000")
conn.execute("PRAGMA synchronous = OFF")


def first_number(value):
    if value is None:
        return ""
    nums = re.findall(r"\d+", str(value))
    return nums[0] if nums else ""


def first_token(value):
    if value is None:
        return ""
    x = str(value).strip()
    return x.split()[0] if x else ""


def last_token(value):
    if value is None:
        return ""
    x = str(value).strip()
    return x.split()[-1] if x else ""


conn.create_function(
    "first_number", 1, first_number, deterministic=True
)

conn.create_function(
    "first_token", 1, first_token, deterministic=True
)

conn.create_function(
    "last_token", 1, last_token, deterministic=True
)


print("=" * 70)
print("ANALYZING MISSED POSITIVE PAIRS")
print("=" * 70)

start = time.time()

# ------------------------------------------------------------
# Create temporary table containing ONLY missed positives
# ------------------------------------------------------------

print("\nFinding missed positive pairs...")

conn.execute("DROP TABLE IF EXISTS missed_pairs")

conn.execute("""
CREATE TEMP TABLE missed_pairs AS
SELECT
    p.source1_entity_id,
    p.matched_entity_id,
    p.matched_source
FROM positive_pairs p
JOIN source1 a
    ON a.entity_id = p.source1_entity_id
JOIN
(
    SELECT
        entity_id,
        business_name_norm,
        business_name_clean,
        address_numbers
    FROM source2
) b
    ON p.matched_source = 'S2'
   AND b.entity_id = p.matched_entity_id
WHERE
    NOT (
        (
            a.business_name_norm != ''
            AND a.business_name_norm = b.business_name_norm
        )
        OR
        (
            a.business_name_norm != ''
            AND substr(a.business_name_norm,1,3)
                = substr(b.business_name_norm,1,3)
        )
        OR
        (
            a.business_name_clean != ''
            AND substr(a.business_name_clean,1,4)
                = substr(b.business_name_clean,1,4)
        )
        OR
        (
            first_number(a.address_numbers) != ''
            AND first_number(a.address_numbers)
                = first_number(b.address_numbers)
        )
    )
""")

# S3 missed pairs
conn.execute("""
INSERT INTO missed_pairs
SELECT
    p.source1_entity_id,
    p.matched_entity_id,
    p.matched_source
FROM positive_pairs p
JOIN source1 a
    ON a.entity_id = p.source1_entity_id
JOIN
(
    SELECT
        entity_id,
        business_name_norm,
        business_name_clean,
        address_numbers
    FROM source3
) b
    ON p.matched_source = 'S3'
   AND b.entity_id = p.matched_entity_id
WHERE
    NOT (
        (
            a.business_name_norm != ''
            AND a.business_name_norm = b.business_name_norm
        )
        OR
        (
            a.business_name_norm != ''
            AND substr(a.business_name_norm,1,3)
                = substr(b.business_name_norm,1,3)
        )
        OR
        (
            a.business_name_clean != ''
            AND substr(a.business_name_clean,1,4)
                = substr(b.business_name_clean,1,4)
        )
        OR
        (
            first_number(a.address_numbers) != ''
            AND first_number(a.address_numbers)
                = first_number(b.address_numbers)
        )
    )
""")

conn.execute("""
CREATE INDEX idx_missed_s1
ON missed_pairs(source1_entity_id)
""")

conn.execute("""
CREATE INDEX idx_missed_candidate
ON missed_pairs(matched_entity_id)
""")

conn.execute("""
CREATE INDEX idx_missed_source
ON missed_pairs(matched_source)
""")

missed = conn.execute(
    "SELECT COUNT(*) FROM missed_pairs"
).fetchone()[0]

print(f"Missed pairs: {missed:,}")


# ============================================================
# Function to evaluate an additional key
# ============================================================

def evaluate_key(name, condition):

    print(f"\nChecking {name}...")

    query = f"""
    SELECT COUNT(*)
    FROM missed_pairs p
    JOIN source1 a
        ON a.entity_id = p.source1_entity_id
    LEFT JOIN source2 b2
        ON p.matched_source = 'S2'
       AND b2.entity_id = p.matched_entity_id
    LEFT JOIN source3 b3
        ON p.matched_source = 'S3'
       AND b3.entity_id = p.matched_entity_id
    WHERE {condition}
    """

    count = conn.execute(query).fetchone()[0]

    print(
        f"{name}: {count:,} "
        f"({count / missed:.4%} of missed)"
    )

    return count


results = {}


# ============================================================
# Additional candidate keys
# ============================================================

results["name_prefix5"] = evaluate_key(
    "Name prefix-5",
    """
    (
        a.business_name_norm != ''
        AND (
            substr(a.business_name_norm,1,5)
            =
            CASE
                WHEN p.matched_source = 'S2'
                THEN substr(b2.business_name_norm,1,5)
                ELSE substr(b3.business_name_norm,1,5)
            END
        )
    )
    """
)


results["clean_prefix5"] = evaluate_key(
    "Clean-name prefix-5",
    """
    (
        a.business_name_clean != ''
        AND (
            substr(a.business_name_clean,1,5)
            =
            CASE
                WHEN p.matched_source = 'S2'
                THEN substr(b2.business_name_clean,1,5)
                ELSE substr(b3.business_name_clean,1,5)
            END
        )
    )
    """
)


results["name_suffix4"] = evaluate_key(
    "Name suffix-4",
    """
    (
        a.business_name_norm != ''
        AND (
            substr(a.business_name_norm,-4)
            =
            CASE
                WHEN p.matched_source = 'S2'
                THEN substr(b2.business_name_norm,-4)
                ELSE substr(b3.business_name_norm,-4)
            END
        )
    )
    """
)


results["first_token"] = evaluate_key(
    "First name token",
    """
    (
        first_token(a.business_name_norm) != ''
        AND (
            first_token(a.business_name_norm)
            =
            CASE
                WHEN p.matched_source = 'S2'
                THEN first_token(b2.business_name_norm)
                ELSE first_token(b3.business_name_norm)
            END
        )
    )
    """
)


results["last_token"] = evaluate_key(
    "Last name token",
    """
    (
        last_token(a.business_name_norm) != ''
        AND (
            last_token(a.business_name_norm)
            =
            CASE
                WHEN p.matched_source = 'S2'
                THEN last_token(b2.business_name_norm)
                ELSE last_token(b3.business_name_norm)
            END
        )
    )
    """
)


results["address_exact"] = evaluate_key(
    "Exact normalized address",
    """
    (
        a.business_address_norm != ''
        AND (
            a.business_address_norm
            =
            CASE
                WHEN p.matched_source = 'S2'
                THEN b2.business_address_norm
                ELSE b3.business_address_norm
            END
        )
    )
    """
)


results["address_clean_exact"] = evaluate_key(
    "Exact cleaned address",
    """
    (
        a.business_address_clean != ''
        AND (
            a.business_address_clean
            =
            CASE
                WHEN p.matched_source = 'S2'
                THEN b2.business_address_clean
                ELSE b3.business_address_clean
            END
        )
    )
    """
)


results["address_prefix8"] = evaluate_key(
    "Address prefix-8",
    """
    (
        a.business_address_norm != ''
        AND (
            substr(a.business_address_norm,1,8)
            =
            CASE
                WHEN p.matched_source = 'S2'
                THEN substr(b2.business_address_norm,1,8)
                ELSE substr(b3.business_address_norm,1,8)
            END
        )
    )
    """
)


results["address_prefix12"] = evaluate_key(
    "Address prefix-12",
    """
    (
        a.business_address_norm != ''
        AND (
            substr(a.business_address_norm,1,12)
            =
            CASE
                WHEN p.matched_source = 'S2'
                THEN substr(b2.business_address_norm,1,12)
                ELSE substr(b3.business_address_norm,1,12)
            END
        )
    )
    """
)


# ============================================================
# Summary
# ============================================================

print("\n" + "=" * 70)
print("MISSED-PAIR ANALYSIS SUMMARY")
print("=" * 70)

for key, count in results.items():
    print(
        f"{key:25s} "
        f"{count:10,} "
        f"({count / missed:.4%})"
    )

print("\nElapsed:",
      round((time.time() - start) / 60, 2),
      "minutes")

conn.close()

print("\nCELL 16 COMPLETE")
```

```python
# ============================================================
# CELL 17 — CANDIDATE VOLUME CHECK
# ============================================================

import sqlite3
import re
import time
import statistics

DB_PATH = (
    "/content/drive/MyDrive/"
    "amazon_training/databases/"
    "training_lookup.db"
)

conn = sqlite3.connect(DB_PATH)

conn.execute("PRAGMA temp_store = FILE")
conn.execute("PRAGMA cache_size = -50000")
conn.execute("PRAGMA synchronous = OFF")


# ------------------------------------------------------------
# Helpers
# ------------------------------------------------------------

def first_number(value):
    if value is None:
        return ""
    nums = re.findall(r"\d+", str(value))
    return nums[0] if nums else ""


conn.create_function(
    "first_number",
    1,
    first_number,
    deterministic=True
)


print("=" * 70)
print("CELL 17 — CANDIDATE VOLUME CHECK")
print("=" * 70)

start = time.time()


# ============================================================
# Create temporary candidate table
# ============================================================

print("\nCreating temporary candidate pairs...")

conn.execute("DROP TABLE IF EXISTS temp_candidates")

conn.execute("""
CREATE TEMP TABLE temp_candidates AS

-- ----------------------------------------------------------
-- KEY 1: EXACT NORMALIZED NAME
-- ----------------------------------------------------------

SELECT
    a.entity_id AS source1_entity_id,
    b.entity_id AS matched_entity_id,
    'S2' AS matched_source
FROM source1 a
JOIN source2 b
    ON a.business_name_norm != ''
   AND a.business_name_norm = b.business_name_norm

UNION

SELECT
    a.entity_id,
    b.entity_id,
    'S3'
FROM source1 a
JOIN source3 b
    ON a.business_name_norm != ''
   AND a.business_name_norm = b.business_name_norm


-- ----------------------------------------------------------
-- KEY 2: NAME PREFIX-3
-- ----------------------------------------------------------

UNION

SELECT
    a.entity_id,
    b.entity_id,
    'S2'
FROM source1 a
JOIN source2 b
    ON a.business_name_norm != ''
   AND substr(a.business_name_norm,1,3)
       = substr(b.business_name_norm,1,3)

UNION

SELECT
    a.entity_id,
    b.entity_id,
    'S3'
FROM source1 a
JOIN source3 b
    ON a.business_name_norm != ''
   AND substr(a.business_name_norm,1,3)
       = substr(b.business_name_norm,1,3)


-- ----------------------------------------------------------
-- KEY 3: CLEAN NAME PREFIX-4
-- ----------------------------------------------------------

UNION

SELECT
    a.entity_id,
    b.entity_id,
    'S2'
FROM source1 a
JOIN source2 b
    ON a.business_name_clean != ''
   AND substr(a.business_name_clean,1,4)
       = substr(b.business_name_clean,1,4)

UNION

SELECT
    a.entity_id,
    b.entity_id,
    'S3'
FROM source1 a
JOIN source3 b
    ON a.business_name_clean != ''
   AND substr(a.business_name_clean,1,4)
       = substr(b.business_name_clean,1,4)


-- ----------------------------------------------------------
-- KEY 4: ADDRESS NUMBER
-- ----------------------------------------------------------

UNION

SELECT
    a.entity_id,
    b.entity_id,
    'S2'
FROM source1 a
JOIN source2 b
    ON first_number(a.address_numbers) != ''
   AND first_number(a.address_numbers)
       = first_number(b.address_numbers)

UNION

SELECT
    a.entity_id,
    b.entity_id,
    'S3'
FROM source1 a
JOIN source3 b
    ON first_number(a.address_numbers) != ''
   AND first_number(a.address_numbers)
       = first_number(b.address_numbers)


-- ----------------------------------------------------------
-- KEY 5: EXACT NORMALIZED ADDRESS
-- ----------------------------------------------------------

UNION

SELECT
    a.entity_id,
    b.entity_id,
    'S2'
FROM source1 a
JOIN source2 b
    ON a.business_address_norm != ''
   AND a.business_address_norm = b.business_address_norm

UNION

SELECT
    a.entity_id,
    b.entity_id,
    'S3'
FROM source1 a
JOIN source3 b
    ON a.business_address_norm != ''
   AND a.business_address_norm = b.business_address_norm
""")


# ============================================================
# BASIC COUNTS
# ============================================================

print("\nCalculating candidate statistics...")

total_candidates = conn.execute("""
    SELECT COUNT(*)
    FROM temp_candidates
""").fetchone()[0]

print(f"\nTotal unique candidates: {total_candidates:,}")


s2_candidates = conn.execute("""
    SELECT COUNT(*)
    FROM temp_candidates
    WHERE matched_source = 'S2'
""").fetchone()[0]

s3_candidates = conn.execute("""
    SELECT COUNT(*)
    FROM temp_candidates
    WHERE matched_source = 'S3'
""").fetchone()[0]

print(f"S2 candidates:          {s2_candidates:,}")
print(f"S3 candidates:          {s3_candidates:,}")


# ============================================================
# S1 COVERAGE
# ============================================================

s1_total = conn.execute("""
    SELECT COUNT(*)
    FROM source1
""").fetchone()[0]

s1_with_candidates = conn.execute("""
    SELECT COUNT(DISTINCT source1_entity_id)
    FROM temp_candidates
""").fetchone()[0]

s1_without = s1_total - s1_with_candidates

print("\nS1 coverage:")
print(f"Total S1 entities:       {s1_total:,}")
print(f"S1 with candidates:      {s1_with_candidates:,}")
print(f"S1 without candidates:   {s1_without:,}")
print(
    f"S1 coverage:             "
    f"{s1_with_candidates / s1_total:.4%}"
)


# ============================================================
# CANDIDATES PER S1
# ============================================================

print("\nCalculating candidates per S1...")

counts = [
    row[0]
    for row in conn.execute("""
        SELECT COUNT(*)
        FROM temp_candidates
        GROUP BY source1_entity_id
    """)
]


# ============================================================
# DISTRIBUTION
# ============================================================

counts.sort()

print("\nCandidate distribution")
print("-" * 50)

print(f"Mean:       {sum(counts) / len(counts):,.2f}")
print(f"Median:     {statistics.median(counts):,.2f}")

# Percentiles
def percentile(values, p):
    if not values:
        return 0

    k = (len(values) - 1) * p
    f = int(k)
    c = min(f + 1, len(values) - 1)

    if f == c:
        return values[f]

    return values[f] + (values[c] - values[f]) * (k - f)


print(f"90th pct:   {percentile(counts, 0.90):,.2f}")
print(f"95th pct:   {percentile(counts, 0.95):,.2f}")
print(f"99th pct:   {percentile(counts, 0.99):,.2f}")
print(f"99.9th pct: {percentile(counts, 0.999):,.2f}")
print(f"Maximum:    {max(counts):,}")


# ============================================================
# LARGE CANDIDATE SETS
# ============================================================

print("\nLarge candidate sets")
print("-" * 50)

for threshold in [100, 500, 1000, 5000, 10000]:

    n = sum(x > threshold for x in counts)

    print(
        f"S1 with > {threshold:>5,} candidates: "
        f"{n:,}"
    )


# ============================================================
# TOP 20 LARGEST S1 CANDIDATE SETS
# ============================================================

print("\nTop 20 largest candidate sets")
print("-" * 50)

top20 = conn.execute("""
    SELECT
        source1_entity_id,
        COUNT(*) AS candidate_count
    FROM temp_candidates
    GROUP BY source1_entity_id
    ORDER BY candidate_count DESC
    LIMIT 20
""").fetchall()

for entity_id, count in top20:
    print(f"{entity_id}\t{count:,}")


# ============================================================
# ESTIMATED FEATURE DATA SIZE
# ============================================================

print("\nEstimated workload")
print("-" * 50)

print(
    f"Candidates to score: {total_candidates:,}"
)

print(
    f"Approx. candidates per S1: "
    f"{total_candidates / s1_total:,.2f}"
)


# ============================================================
# FINISH
# ============================================================

elapsed = (time.time() - start) / 60

print("\n" + "=" * 70)
print("CELL 17 COMPLETE")
print("=" * 70)

print(f"Elapsed time: {elapsed:.2f} minutes")

conn.close()
```

```python
# ============================================================
# CELL 18E — BUILD PERSISTENT TEST DATABASE
# ============================================================

import sqlite3
import pandas as pd
import os
import time

BASE = "/content/drive/MyDrive/amazon_training"

TEST_DIR = BASE + "/test_prepared"
DB_PATH = BASE + "/databases/test_lookup.db"

FILES = {
    "test_source1": TEST_DIR + "/test_s1_prepared.tsv",
    "test_source2": TEST_DIR + "/test_s2_prepared.tsv",
    "test_source3": TEST_DIR + "/test_s3_prepared.tsv"
}

print("=" * 70)
print("CELL 18E — BUILDING TEST DATABASE")
print("=" * 70)

print("\nDatabase:")
print(DB_PATH)

conn = sqlite3.connect(DB_PATH)

conn.execute("PRAGMA journal_mode=WAL")
conn.execute("PRAGMA synchronous=NORMAL")
conn.execute("PRAGMA temp_store=FILE")
conn.execute("PRAGMA cache_size=-50000")

# ------------------------------------------------------------
# Check existing tables
# ------------------------------------------------------------

existing = {
    row[0]
    for row in conn.execute("""
        SELECT name
        FROM sqlite_master
        WHERE type='table'
    """)
}

print("\nExisting tables:", existing)


# ------------------------------------------------------------
# Import each test file
# ------------------------------------------------------------

for table, path in FILES.items():

    if table in existing:

        count = conn.execute(
            f"SELECT COUNT(*) FROM {table}"
        ).fetchone()[0]

        print(
            f"\n{table} already exists: "
            f"{count:,} rows — SKIPPING"
        )
        continue

    print(f"\nImporting {table}...")
    print(path)

    start = time.time()

    first = True
    total = 0

    for chunk in pd.read_csv(
        path,
        sep="\t",
        dtype=str,
        chunksize=100_000
    ):

        # Keep only the columns needed for blocking/matching
        columns = [
            "entity_id",
            "country_norm",
            "business_name_norm",
            "business_address_norm",
            "business_name_clean",
            "business_address_clean",
            "address_numbers"
        ]

        chunk = chunk[columns]

        chunk.to_sql(
            table,
            conn,
            if_exists="replace" if first else "append",
            index=False
        )

        total += len(chunk)
        first = False

        print(
            f"  Imported: {total:,} rows | "
            f"{(time.time() - start)/60:.1f} min"
        )

    conn.commit()

    print(
        f"Completed {table}: "
        f"{total:,} rows"
    )


# ------------------------------------------------------------
# Create entity ID indexes
# ------------------------------------------------------------

print("\nCreating entity ID indexes...")

for table in FILES:

    index_name = f"idx_{table}_entity"

    conn.execute(f"""
        CREATE INDEX IF NOT EXISTS
        {index_name}
        ON {table}(entity_id)
    """)

    conn.commit()

print("Entity indexes complete.")


# ------------------------------------------------------------
# Create blocking indexes
# ------------------------------------------------------------

print("\nCreating blocking indexes...")

index_specs = [

    # S1
    (
        "idx_test_s1_name",
        "test_source1",
        "business_name_norm"
    ),
    (
        "idx_test_s1_clean",
        "test_source1",
        "business_name_clean"
    ),
    (
        "idx_test_s1_address",
        "test_source1",
        "business_address_norm"
    ),
    (
        "idx_test_s1_number",
        "test_source1",
        "address_numbers"
    ),

    # S2
    (
        "idx_test_s2_name",
        "test_source2",
        "business_name_norm"
    ),
    (
        "idx_test_s2_clean",
        "test_source2",
        "business_name_clean"
    ),
    (
        "idx_test_s2_address",
        "test_source2",
        "business_address_norm"
    ),
    (
        "idx_test_s2_number",
        "test_source2",
        "address_numbers"
    ),

    # S3
    (
        "idx_test_s3_name",
        "test_source3",
        "business_name_norm"
    ),
    (
        "idx_test_s3_clean",
        "test_source3",
        "business_name_clean"
    ),
    (
        "idx_test_s3_address",
        "test_source3",
        "business_address_norm"
    ),
    (
        "idx_test_s3_number",
        "test_source3",
        "address_numbers"
    )
]

for i, (idx, table, column) in enumerate(index_specs, 1):

    exists = conn.execute("""
        SELECT COUNT(*)
        FROM sqlite_master
        WHERE type='index'
          AND name=?
    """, (idx,)).fetchone()[0]

    if exists:
        print(
            f"[{i}/{len(index_specs)}] "
            f"{idx} already exists ✅"
        )
        continue

    print(
        f"\n[{i}/{len(index_specs)}] "
        f"Creating {idx}..."
    )

    start = time.time()

    conn.execute(f"""
        CREATE INDEX {idx}
        ON {table}({column})
    """)

    conn.commit()

    print(
        f"Completed in "
        f"{(time.time()-start)/60:.2f} min"
    )


# ------------------------------------------------------------
# Final status
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("TEST DATABASE STATUS")
print("=" * 70)

for table in FILES:

    count = conn.execute(
        f"SELECT COUNT(*) FROM {table}"
    ).fetchone()[0]

    print(f"{table}: {count:,}")

index_count = conn.execute("""
    SELECT COUNT(*)
    FROM sqlite_master
    WHERE type='index'
      AND name LIKE 'idx_test_%'
""").fetchone()[0]

print(f"\nTest indexes: {index_count}")

print(
    f"Database size: "
    f"{os.path.getsize(DB_PATH)/(1024**3):.2f} GB"
)

conn.close()

print("\nCELL 18E COMPLETE")
```

```python
# ============================================================
# CELL 18F — EXACT NAME CANDIDATE GENERATION (FIXED)
# ============================================================

import duckdb
import os
import time

BASE = "/content/drive/MyDrive/amazon_training"

TEST_DIR = BASE + "/test_prepared"
CAND_DIR = BASE + "/candidates"

os.makedirs(CAND_DIR, exist_ok=True)
os.makedirs("/content/duckdb_tmp", exist_ok=True)

S1 = TEST_DIR + "/test_s1_prepared.tsv"
S2 = TEST_DIR + "/test_s2_prepared.tsv"
S3 = TEST_DIR + "/test_s3_prepared.tsv"

OUT_S2 = CAND_DIR + "/exact_name_s2.parquet"
OUT_S3 = CAND_DIR + "/exact_name_s3.parquet"

print("=" * 70)
print("CELL 18F — EXACT NAME CANDIDATE GENERATION")
print("=" * 70)

con = duckdb.connect()

con.execute("SET threads=2")
con.execute("SET memory_limit='7GB'")
con.execute("SET temp_directory='/content/duckdb_tmp'")
con.execute("SET preserve_insertion_order=false")

# ------------------------------------------------------------
# S2
# ------------------------------------------------------------

if os.path.exists(OUT_S2) and os.path.getsize(OUT_S2) > 0:

    print("\nS2 output already exists — SKIPPING")

else:

    print("\n[S2] S1 → S2 exact-name blocking")
    start = time.time()

    con.execute(f"""
    COPY (
        SELECT
            s1.entity_id AS source1_entity_id,
            s2.entity_id AS matched_entity_id,
            'S2' AS matched_source
        FROM read_csv(
            '{S1}',
            delim='\\t',
            header=true,
            auto_detect=true,
            strict_mode=false
        ) AS s1
        INNER JOIN read_csv(
            '{S2}',
            delim='\\t',
            header=true,
            auto_detect=true,
            strict_mode=false
        ) AS s2
        ON s1.business_name_norm = s2.business_name_norm
        WHERE s1.business_name_norm <> ''
          AND s2.business_name_norm <> ''
    )
    TO '{OUT_S2}'
    (FORMAT PARQUET, COMPRESSION ZSTD)
    """)

    print(
        f"S2 completed in "
        f"{(time.time()-start)/60:.2f} minutes"
    )


# ------------------------------------------------------------
# S3
# ------------------------------------------------------------

if os.path.exists(OUT_S3) and os.path.getsize(OUT_S3) > 0:

    print("\nS3 output already exists — SKIPPING")

else:

    print("\n[S3] S1 → S3 exact-name blocking")
    start = time.time()

    con.execute(f"""
    COPY (
        SELECT
            s1.entity_id AS source1_entity_id,
            s3.entity_id AS matched_entity_id,
            'S3' AS matched_source
        FROM read_csv(
            '{S1}',
            delim='\\t',
            header=true,
            auto_detect=true,
            strict_mode=false
        ) AS s1
        INNER JOIN read_csv(
            '{S3}',
            delim='\\t',
            header=true,
            auto_detect=true,
            strict_mode=false
        ) AS s3
        ON s1.business_name_norm = s3.business_name_norm
        WHERE s1.business_name_norm <> ''
          AND s3.business_name_norm <> ''
    )
    TO '{OUT_S3}'
    (FORMAT PARQUET, COMPRESSION ZSTD)
    """)

    print(
        f"S3 completed in "
        f"{(time.time()-start)/60:.2f} minutes"
    )


# ------------------------------------------------------------
# Results
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("EXACT NAME RESULTS")
print("=" * 70)

s2_count = con.execute(f"""
    SELECT COUNT(*)
    FROM read_parquet('{OUT_S2}')
""").fetchone()[0]

s3_count = con.execute(f"""
    SELECT COUNT(*)
    FROM read_parquet('{OUT_S3}')
""").fetchone()[0]

print(f"S2 candidates: {s2_count:,}")
print(f"S3 candidates: {s3_count:,}")
print(f"Total candidates: {s2_count + s3_count:,}")

print("\nOutput sizes:")

print(
    f"S2: {os.path.getsize(OUT_S2)/(1024**3):.3f} GB"
)

print(
    f"S3: {os.path.getsize(OUT_S3)/(1024**3):.3f} GB"
)

con.close()

print("\nCELL 18F COMPLETE")
```

```python
# ============================================================
# CELL 18G — EXACT ADDRESS CANDIDATE ANALYSIS
# ============================================================

import duckdb
import os
import time

BASE = "/content/drive/MyDrive/amazon_training"

TEST_DIR = BASE + "/test_prepared"
CAND_DIR = BASE + "/candidates"

S1 = TEST_DIR + "/test_s1_prepared.tsv"
S2 = TEST_DIR + "/test_s2_prepared.tsv"
S3 = TEST_DIR + "/test_s3_prepared.tsv"

NAME_S2 = CAND_DIR + "/exact_name_s2.parquet"
NAME_S3 = CAND_DIR + "/exact_name_s3.parquet"

TMP_DIR = "/content/duckdb_tmp"
os.makedirs(TMP_DIR, exist_ok=True)

con = duckdb.connect()

con.execute("SET threads=2")
con.execute("SET memory_limit='7GB'")
con.execute(f"SET temp_directory='{TMP_DIR}'")
con.execute("SET preserve_insertion_order=false")

print("=" * 70)
print("CELL 18G — EXACT ADDRESS CANDIDATE ANALYSIS")
print("=" * 70)

# ------------------------------------------------------------
# Create temporary exact-name candidate relation
# ------------------------------------------------------------

con.execute(f"""
CREATE OR REPLACE TEMP TABLE name_candidates AS

SELECT
    source1_entity_id,
    matched_entity_id,
    matched_source
FROM read_parquet('{NAME_S2}')

UNION ALL

SELECT
    source1_entity_id,
    matched_entity_id,
    matched_source
FROM read_parquet('{NAME_S3}')
""")

name_count = con.execute("""
SELECT COUNT(*)
FROM name_candidates
""").fetchone()[0]

print(f"\nExisting exact-name candidates: {name_count:,}")


# ------------------------------------------------------------
# Exact address — S2
# ------------------------------------------------------------

print("\n[S2] Measuring exact-address candidates...")
start = time.time()

address_s2 = con.execute(f"""
SELECT COUNT(*)
FROM
(
    SELECT
        s1.entity_id AS source1_entity_id,
        s2.entity_id AS matched_entity_id
    FROM read_csv(
        '{S1}',
        delim='\\t',
        header=true,
        auto_detect=true,
        strict_mode=false
    ) s1
    JOIN read_csv(
        '{S2}',
        delim='\\t',
        header=true,
        auto_detect=true,
        strict_mode=false
    ) s2
      ON s1.business_address_norm = s2.business_address_norm
    WHERE s1.business_address_norm <> ''
      AND s2.business_address_norm <> ''
) x
""").fetchone()[0]

print(f"S2 exact-address candidates: {address_s2:,}")
print(f"Time: {(time.time()-start)/60:.2f} min")


# ------------------------------------------------------------
# Exact address — S3
# ------------------------------------------------------------

print("\n[S3] Measuring exact-address candidates...")
start = time.time()

address_s3 = con.execute(f"""
SELECT COUNT(*)
FROM
(
    SELECT
        s1.entity_id AS source1_entity_id,
        s3.entity_id AS matched_entity_id
    FROM read_csv(
        '{S1}',
        delim='\\t',
        header=true,
        auto_detect=true,
        strict_mode=false
    ) s1
    JOIN read_csv(
        '{S3}',
        delim='\\t',
        header=true,
        auto_detect=true,
        strict_mode=false
    ) s3
      ON s1.business_address_norm = s3.business_address_norm
    WHERE s1.business_address_norm <> ''
      AND s3.business_address_norm <> ''
) x
""").fetchone()[0]

print(f"S3 exact-address candidates: {address_s3:,}")
print(f"Time: {(time.time()-start)/60:.2f} min")


# ------------------------------------------------------------
# Materialize address candidates temporarily
# ------------------------------------------------------------

print("\nCreating temporary address candidate tables...")

con.execute(f"""
CREATE OR REPLACE TEMP TABLE address_candidates AS

SELECT
    s1.entity_id AS source1_entity_id,
    s2.entity_id AS matched_entity_id,
    'S2' AS matched_source
FROM read_csv(
    '{S1}',
    delim='\\t',
    header=true,
    auto_detect=true,
    strict_mode=false
) s1
JOIN read_csv(
    '{S2}',
    delim='\\t',
    header=true,
    auto_detect=true,
    strict_mode=false
) s2
  ON s1.business_address_norm = s2.business_address_norm
WHERE s1.business_address_norm <> ''
  AND s2.business_address_norm <> ''

UNION ALL

SELECT
    s1.entity_id AS source1_entity_id,
    s3.entity_id AS matched_entity_id,
    'S3' AS matched_source
FROM read_csv(
    '{S1}',
    delim='\\t',
    header=true,
    auto_detect=true,
    strict_mode=false
) s1
JOIN read_csv(
    '{S3}',
    delim='\\t',
    header=true,
    auto_detect=true,
    strict_mode=false
) s3
  ON s1.business_address_norm = s3.business_address_norm
WHERE s1.business_address_norm <> ''
  AND s3.business_address_norm <> ''
""")

total_address = con.execute("""
SELECT COUNT(*)
FROM address_candidates
""").fetchone()[0]

print(f"\nTotal exact-address candidates: {total_address:,}")


# ------------------------------------------------------------
# Overlap
# ------------------------------------------------------------

overlap = con.execute("""
SELECT COUNT(*)
FROM address_candidates a
INNER JOIN name_candidates n
    ON a.source1_entity_id = n.source1_entity_id
   AND a.matched_entity_id = n.matched_entity_id
   AND a.matched_source = n.matched_source
""").fetchone()[0]

new_address = total_address - overlap

print("\n" + "=" * 70)
print("OVERLAP ANALYSIS")
print("=" * 70)

print(f"Exact-name candidates:       {name_count:,}")
print(f"Exact-address candidates:    {total_address:,}")
print(f"Overlap:                      {overlap:,}")
print(f"NEW address candidates:       {new_address:,}")

if total_address > 0:
    print(
        f"Address overlap with name:   "
        f"{overlap / total_address * 100:.2f}%"
    )

print("\nCELL 18G COMPLETE")

con.close()
```

```python
# ============================================================
# CELL 18I — TRAINING RECALL OF CURRENT BLOCKING STRATEGIES
# FIXED FOR YOUR DATABASE SCHEMA
# ============================================================

import sqlite3
import time

BASE = "/content/drive/MyDrive/amazon_training"
DB_PATH = BASE + "/databases/training_lookup.db"

print("=" * 70)
print("CELL 18I — TRAINING BLOCKING RECALL")
print("=" * 70)

conn = sqlite3.connect(DB_PATH)

conn.execute("PRAGMA cache_size=-100000")
conn.execute("PRAGMA temp_store=FILE")

# ------------------------------------------------------------
# Show actual positive_pairs schema
# ------------------------------------------------------------

print("\npositive_pairs schema:")

schema = conn.execute("""
PRAGMA table_info(positive_pairs)
""").fetchall()

for row in schema:
    print(f"  {row[1]} : {row[2]}")

# ------------------------------------------------------------
# S2
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("S2 POSITIVE PAIRS")
print("-" * 70)

start = time.time()

s2 = conn.execute("""
SELECT
    COUNT(*) AS total,

    SUM(
        CASE
        WHEN s1.business_name_norm = s2.business_name_norm
        THEN 1 ELSE 0 END
    ) AS exact_name,

    SUM(
        CASE
        WHEN s1.business_name_clean = s2.business_name_clean
        THEN 1 ELSE 0 END
    ) AS clean_name,

    SUM(
        CASE
        WHEN s1.business_address_norm = s2.business_address_norm
             AND s1.business_address_norm <> ''
        THEN 1 ELSE 0 END
    ) AS exact_address,

    SUM(
        CASE
        WHEN
            s1.business_name_norm = s2.business_name_norm
            OR
            s1.business_name_clean = s2.business_name_clean
            OR
            (
                s1.business_address_norm = s2.business_address_norm
                AND s1.business_address_norm <> ''
            )
        THEN 1 ELSE 0 END
    ) AS current_union

FROM positive_pairs p

JOIN source1 s1
    ON p.source1_entity_id = s1.entity_id

JOIN source2 s2
    ON p.matched_entity_id = s2.entity_id
""").fetchone()

print(f"Total S2 positives:       {s2[0]:,}")
print(f"Exact name:               {s2[1]:,}")
print(f"Clean name:               {s2[2]:,}")
print(f"Exact address:            {s2[3]:,}")
print(f"Current union:            {s2[4]:,}")

print(
    f"Time: {(time.time()-start)/60:.2f} min"
)

# ------------------------------------------------------------
# S3
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("S3 POSITIVE PAIRS")
print("-" * 70)

start = time.time()

s3 = conn.execute("""
SELECT
    COUNT(*) AS total,

    SUM(
        CASE
        WHEN s1.business_name_norm = s3.business_name_norm
        THEN 1 ELSE 0 END
    ) AS exact_name,

    SUM(
        CASE
        WHEN s1.business_name_clean = s3.business_name_clean
        THEN 1 ELSE 0 END
    ) AS clean_name,

    SUM(
        CASE
        WHEN s1.business_address_norm = s3.business_address_norm
             AND s1.business_address_norm <> ''
        THEN 1 ELSE 0 END
    ) AS exact_address,

    SUM(
        CASE
        WHEN
            s1.business_name_norm = s3.business_name_norm
            OR
            s1.business_name_clean = s3.business_name_clean
            OR
            (
                s1.business_address_norm = s3.business_address_norm
                AND s1.business_address_norm <> ''
            )
        THEN 1 ELSE 0 END
    ) AS current_union

FROM positive_pairs p

JOIN source1 s1
    ON p.source1_entity_id = s1.entity_id

JOIN source3 s3
    ON p.matched_entity_id = s3.entity_id
""").fetchone()

print(f"Total S3 positives:       {s3[0]:,}")
print(f"Exact name:               {s3[1]:,}")
print(f"Clean name:               {s3[2]:,}")
print(f"Exact address:            {s3[3]:,}")
print(f"Current union:            {s3[4]:,}")

print(
    f"Time: {(time.time()-start)/60:.2f} min"
)

# ------------------------------------------------------------
# Overall
# ------------------------------------------------------------

total_positive = s2[0] + s3[0]
current_union = s2[4] + s3[4]
missed = total_positive - current_union

recall = current_union / total_positive

print("\n" + "=" * 70)
print("CURRENT BLOCKING RECALL")
print("=" * 70)

print(f"Total positive pairs:     {total_positive:,}")
print(f"Found by current blocks:  {current_union:,}")
print(f"Missed:                   {missed:,}")
print(f"Recall:                   {recall * 100:.4f}%")

conn.close()

print("\nCELL 18I COMPLETE")
```

```python
import duckdb
import time

BASE = "/content/drive/MyDrive/amazon_training"
DB = f"{BASE}/databases/test_candidates.duckdb"
CAND = f"{BASE}/candidates/final_candidate_union.parquet"

con = duckdb.connect(DB)

con.execute("PRAGMA threads=2")
con.execute("PRAGMA memory_limit='4GB'")

print("=" * 70)
print("CELL 19F — FINAL CANDIDATE DISTRIBUTION CHECK")
print("=" * 70)

start = time.time()

# ------------------------------------------------------------
# Overall counts
# ------------------------------------------------------------
overall = con.execute(f"""
    SELECT
        COUNT(*) AS total_candidates,
        COUNT(DISTINCT source1_entity_id) AS source1_with_candidates
    FROM read_parquet('{CAND}')
""").fetchone()

print("\nOVERALL")
print("-" * 70)
print(f"Total candidates       : {overall[0]:,}")
print(f"S1 entities covered    : {overall[1]:,}")

# ------------------------------------------------------------
# Candidates per S1 entity
# ------------------------------------------------------------
stats = con.execute(f"""
    WITH counts AS (
        SELECT
            source1_entity_id,
            COUNT(*) AS candidate_count
        FROM read_parquet('{CAND}')
        GROUP BY source1_entity_id
    )
    SELECT
        COUNT(*) AS entities,
        MIN(candidate_count) AS min_candidates,
        ROUND(AVG(candidate_count), 2) AS avg_candidates,
        MEDIAN(candidate_count) AS median_candidates,
        MAX(candidate_count) AS max_candidates,
        SUM(CASE WHEN candidate_count = 0 THEN 1 ELSE 0 END)
            AS zero_candidate_entities,
        SUM(CASE WHEN candidate_count > 100 THEN 1 ELSE 0 END)
            AS over_100,
        SUM(CASE WHEN candidate_count > 500 THEN 1 ELSE 0 END)
            AS over_500,
        SUM(CASE WHEN candidate_count > 1000 THEN 1 ELSE 0 END)
            AS over_1000
    FROM counts
""").fetchone()

print("\nCANDIDATES PER SOURCE-1 ENTITY")
print("-" * 70)
print(f"Entities with candidates : {stats[0]:,}")
print(f"Minimum                 : {stats[1]:,}")
print(f"Average                 : {stats[2]:,.2f}")
print(f"Median                  : {stats[3]:,.0f}")
print(f"Maximum                 : {stats[4]:,}")
print(f"Zero candidates         : {stats[5]:,}")
print(f"> 100 candidates        : {stats[6]:,}")
print(f"> 500 candidates        : {stats[7]:,}")
print(f"> 1000 candidates       : {stats[8]:,}")

# ------------------------------------------------------------
# S2 / S3 distribution
# ------------------------------------------------------------
source_stats = con.execute(f"""
    SELECT
        matched_source,
        COUNT(*) AS candidates,
        COUNT(DISTINCT source1_entity_id) AS s1_entities
    FROM read_parquet('{CAND}')
    GROUP BY matched_source
    ORDER BY matched_source
""").fetchall()

print("\nSOURCE DISTRIBUTION")
print("-" * 70)

for row in source_stats:
    print(
        f"{row[0]:10s} : "
        f"{row[1]:,} candidates | "
        f"{row[2]:,} S1 entities"
    )

elapsed = (time.time() - start) / 60

print("\n" + "=" * 70)
print(f"Check time: {elapsed:.2f} min")
print("CELL 19F COMPLETE")
print("=" * 70)

con.close()
```

```python
import duckdb
import os
import time

BASE = "/content/drive/MyDrive/amazon_training"
DB = f"{BASE}/databases/test_candidates.duckdb"
CAND = f"{BASE}/candidates"

S1 = f"{BASE}/test_prepared/test_s1_prepared.tsv"
S2 = f"{BASE}/test_prepared/test_s2_prepared.tsv"
S3 = f"{BASE}/test_prepared/test_s3_prepared.tsv"

os.makedirs(CAND, exist_ok=True)

con = duckdb.connect(DB)
con.execute("PRAGMA threads=2")
con.execute("PRAGMA memory_limit='4GB'")

print("=" * 70)
print("CELL 20D — FINISH BLOCKING + BUILD FINAL UNION")
print("=" * 70)

overall_start = time.time()


# ============================================================
# 1. S1 → S3 CLEAN NAME
# ============================================================

print("\n[1/4] S1 → S3 exact-clean-name...")

start = time.time()

out = f"{CAND}/exact_clean_name_s3.parquet"

con.execute(f"""
COPY (
    SELECT
        s1.entity_id AS source1_entity_id,
        s3.entity_id AS matched_entity_id,
        'source3' AS matched_source
    FROM read_csv(
        '{S1}',
        header=true,
        auto_detect=true,
        strict_mode=false
    ) s1
    INNER JOIN read_csv(
        '{S3}',
        header=true,
        auto_detect=true,
        strict_mode=false
    ) s3
    ON s1.country_norm = s3.country_norm
    AND s1.business_name_clean = s3.business_name_clean
    AND s1.business_name_clean IS NOT NULL
    AND s1.business_name_clean != ''
)
TO '{out}'
(FORMAT PARQUET, COMPRESSION ZSTD);
""")

count = con.execute(
    f"SELECT COUNT(*) FROM read_parquet('{out}')"
).fetchone()[0]

s1_count = con.execute(
    f"SELECT COUNT(DISTINCT source1_entity_id) FROM read_parquet('{out}')"
).fetchone()[0]

print(f"    Candidates   : {count:,}")
print(f"    S1 entities  : {s1_count:,}")
print(f"    Time         : {(time.time()-start)/60:.2f} min")


# ============================================================
# 2. S1 → S2 EXACT ADDRESS
# ============================================================

print("\n[2/4] S1 → S2 exact-address...")

start = time.time()

out = f"{CAND}/exact_address_s2.parquet"

con.execute(f"""
COPY (
    SELECT
        s1.entity_id AS source1_entity_id,
        s2.entity_id AS matched_entity_id,
        'source2' AS matched_source
    FROM read_csv(
        '{S1}',
        header=true,
        auto_detect=true,
        strict_mode=false
    ) s1
    INNER JOIN read_csv(
        '{S2}',
        header=true,
        auto_detect=true,
        strict_mode=false
    ) s2
    ON s1.country_norm = s2.country_norm
    AND s1.business_address_norm = s2.business_address_norm
    AND s1.business_address_norm IS NOT NULL
    AND s1.business_address_norm != ''
)
TO '{out}'
(FORMAT PARQUET, COMPRESSION ZSTD);
""")

count = con.execute(
    f"SELECT COUNT(*) FROM read_parquet('{out}')"
).fetchone()[0]

s1_count = con.execute(
    f"SELECT COUNT(DISTINCT source1_entity_id) FROM read_parquet('{out}')"
).fetchone()[0]

print(f"    Candidates   : {count:,}")
print(f"    S1 entities  : {s1_count:,}")
print(f"    Time         : {(time.time()-start)/60:.2f} min")


# ============================================================
# 3. S1 → S3 EXACT ADDRESS
# ============================================================

print("\n[3/4] S1 → S3 exact-address...")

start = time.time()

out = f"{CAND}/exact_address_s3.parquet"

con.execute(f"""
COPY (
    SELECT
        s1.entity_id AS source1_entity_id,
        s3.entity_id AS matched_entity_id,
        'source3' AS matched_source
    FROM read_csv(
        '{S1}',
        header=true,
        auto_detect=true,
        strict_mode=false
    ) s1
    INNER JOIN read_csv(
        '{S3}',
        header=true,
        auto_detect=true,
        strict_mode=false
    ) s3
    ON s1.country_norm = s3.country_norm
    AND s1.business_address_norm = s3.business_address_norm
    AND s1.business_address_norm IS NOT NULL
    AND s1.business_address_norm != ''
)
TO '{out}'
(FORMAT PARQUET, COMPRESSION ZSTD);
""")

count = con.execute(
    f"SELECT COUNT(*) FROM read_parquet('{out}')"
).fetchone()[0]

s1_count = con.execute(
    f"SELECT COUNT(DISTINCT source1_entity_id) FROM read_parquet('{out}')"
).fetchone()[0]

print(f"    Candidates   : {count:,}")
print(f"    S1 entities  : {s1_count:,}")
print(f"    Time         : {(time.time()-start)/60:.2f} min")


# ============================================================
# 4. FINAL UNION
# ============================================================

print("\n[4/4] Building final deduplicated union...")

start = time.time()

final_out = f"{CAND}/final_candidate_union.parquet"

con.execute(f"""
COPY (
    SELECT DISTINCT
        source1_entity_id,
        matched_entity_id,
        matched_source
    FROM (
        SELECT * FROM read_parquet('{CAND}/exact_name_s2.parquet')
        UNION ALL
        SELECT * FROM read_parquet('{CAND}/exact_name_s3.parquet')
        UNION ALL
        SELECT * FROM read_parquet('{CAND}/exact_clean_name_s2.parquet')
        UNION ALL
        SELECT * FROM read_parquet('{CAND}/exact_clean_name_s3.parquet')
        UNION ALL
        SELECT * FROM read_parquet('{CAND}/exact_address_s2.parquet')
        UNION ALL
        SELECT * FROM read_parquet('{CAND}/exact_address_s3.parquet')
    )
)
TO '{final_out}'
(FORMAT PARQUET, COMPRESSION ZSTD);
""")


# ============================================================
# FINAL VALIDATION
# ============================================================

total = con.execute(f"""
    SELECT COUNT(*)
    FROM read_parquet('{final_out}')
""").fetchone()[0]

distinct_s1 = con.execute(f"""
    SELECT COUNT(DISTINCT source1_entity_id)
    FROM read_parquet('{final_out}')
""").fetchone()[0]

s2_total = con.execute(f"""
    SELECT COUNT(*)
    FROM read_parquet('{final_out}')
    WHERE matched_source = 'source2'
""").fetchone()[0]

s3_total = con.execute(f"""
    SELECT COUNT(*)
    FROM read_parquet('{final_out}')
    WHERE matched_source = 'source3'
""").fetchone()[0]

distribution = con.execute(f"""
    WITH counts AS (
        SELECT
            source1_entity_id,
            COUNT(*) AS n
        FROM read_parquet('{final_out}')
        GROUP BY source1_entity_id
    )
    SELECT
        MIN(n),
        ROUND(AVG(n), 2),
        MEDIAN(n),
        MAX(n),
        SUM(CASE WHEN n > 100 THEN 1 ELSE 0 END)
    FROM counts
""").fetchone()

size_gb = os.path.getsize(final_out) / (1024**3)

print("\n" + "=" * 70)
print("FINAL CANDIDATE UNION")
print("=" * 70)

print(f"Total candidates       : {total:,}")
print(f"S2 candidates          : {s2_total:,}")
print(f"S3 candidates          : {s3_total:,}")
print(f"Distinct Test S1       : {distinct_s1:,}")
print(f"Min candidates/S1     : {distribution[0]:,}")
print(f"Average candidates/S1  : {distribution[1]:,.2f}")
print(f"Median candidates/S1   : {distribution[2]:,.0f}")
print(f"Maximum candidates/S1  : {distribution[3]:,}")
print(f"S1 with >100          : {distribution[4]:,}")
print(f"Final Parquet size     : {size_gb:.3f} GB")

print(f"\nTotal time: {(time.time()-overall_start)/60:.2f} min")

con.close()

print("\nCELL 20D COMPLETE")
```

```python
# ============================================================
# LOCATE THE ORIGINAL FEATURE-GENERATION CODE
# ============================================================

import os
from pathlib import Path

roots = [
    Path("/content/drive/MyDrive/amazon_training"),
    Path("/content/drive/MyDrive"),
]

keywords = [
    "name_token_sort",
    "number_overlap",
    "address_token_set",
    "training_features_100k",
]

extensions = {".py", ".ipynb", ".txt", ".md"}

found = []

for root in roots:
    if not root.exists():
        continue

    for path in root.rglob("*"):
        if not path.is_file():
            continue

        if path.suffix.lower() not in extensions:
            continue

        # Skip huge/non-code artifacts
        try:
            if path.stat().st_size > 20 * 1024 * 1024:
                continue

            text = path.read_text(errors="ignore")
        except Exception:
            continue

        hits = [k for k in keywords if k in text]

        if hits:
            found.append((str(path), hits))

print("=" * 80)
print("FEATURE CODE SEARCH")
print("=" * 80)

if not found:
    print("No feature-generation source code found.")
else:
    for path, hits in found:
        print("\nFILE:", path)
        print("MATCHES:", ", ".join(hits))
```

```python
# ============================================================
# STEP 2 — SCORE TEST S1 -> S2 CANDIDATES
# ============================================================

import os
import gc
import time
import duckdb
import numpy as np
import pandas as pd
import pyarrow as pa
import pyarrow.parquet as pq
import lightgbm as lgb

from rapidfuzz.fuzz import (
    ratio,
    token_sort_ratio,
    token_set_ratio
)

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

BASE = "/content/drive/MyDrive/amazon_training"

CAND_PATH = BASE + "/candidates/final_candidate_union.parquet"
S1_PATH = BASE + "/test_prepared/test_s1_prepared.tsv"
S2_PATH = BASE + "/test_prepared/test_s2_prepared.tsv"

SCORES_DIR = BASE + "/features"
OUTPUT_PATH = SCORES_DIR + "/test_scores_s2.parquet"

MODEL_PATH = BASE + "/models/entity_match_lightgbm.txt"

os.makedirs(SCORES_DIR, exist_ok=True)

# ------------------------------------------------------------
# FEATURES — EXACT TRAINING ORDER
# ------------------------------------------------------------

FEATURES = [
    "name_exact",
    "name_clean_exact",
    "name_ratio",
    "name_token_sort",
    "name_token_set",
    "address_exact",
    "address_clean_exact",
    "address_ratio",
    "address_token_sort",
    "address_token_set",
    "number_overlap",
    "name_length_diff",
    "address_length_diff",
    "country_match",
    "source_type"
]

# ------------------------------------------------------------
# RESUME CHECK
# ------------------------------------------------------------

if os.path.exists(OUTPUT_PATH):
    print("=" * 80)
    print("S2 SCORES ALREADY EXIST")
    print("=" * 80)
    print(OUTPUT_PATH)

    con = duckdb.connect()
    result = con.execute(f"""
        SELECT
            COUNT(*) AS rows,
            COUNT(DISTINCT source1_entity_id) AS s1_entities,
            MIN(score) AS min_score,
            MAX(score) AS max_score,
            AVG(score) AS avg_score
        FROM read_parquet('{OUTPUT_PATH}')
    """).fetchone()

    print(f"Rows:        {result[0]:,}")
    print(f"S1 entities: {result[1]:,}")
    print(f"Min score:   {result[2]:.6f}")
    print(f"Max score:   {result[3]:.6f}")
    print(f"Avg score:   {result[4]:.6f}")

    con.close()

else:

    print("=" * 80)
    print("STEP 2 — SCORING S1 -> S2")
    print("=" * 80)

    start_time = time.time()

    # --------------------------------------------------------
    # Load model
    # --------------------------------------------------------

    model = lgb.Booster(model_file=MODEL_PATH)

    print("Model loaded.")
    print("Trees:", model.num_trees())

    # --------------------------------------------------------
    # DuckDB
    # --------------------------------------------------------

    tmp_dir = "/content/duckdb_tmp"
    os.makedirs(tmp_dir, exist_ok=True)

    con = duckdb.connect()

    con.execute("SET threads=2")
    con.execute("SET memory_limit='5GB'")
    con.execute(
        f"SET temp_directory='{tmp_dir}'"
    )
    con.execute(
        "SET preserve_insertion_order=false"
    )

    # --------------------------------------------------------
    # Create temporary source tables
    # --------------------------------------------------------

    print("\nLoading S1 and S2 into DuckDB...")

    con.execute(f"""
        CREATE OR REPLACE TEMP TABLE s1 AS
        SELECT
            entity_id,
            country_norm,
            business_name_norm,
            business_address_norm,
            business_name_clean,
            business_address_clean
        FROM read_csv(
            '{S1_PATH}',
            delim='\\t',
            header=true,
            auto_detect=true,
            strict_mode=false
        )
    """)

    con.execute(f"""
        CREATE OR REPLACE TEMP TABLE s2 AS
        SELECT
            entity_id,
            country_norm,
            business_name_norm,
            business_address_norm,
            business_name_clean,
            business_address_clean
        FROM read_csv(
            '{S2_PATH}',
            delim='\\t',
            header=true,
            auto_detect=true,
            strict_mode=false
        )
    """)

    s1_count = con.execute(
        "SELECT COUNT(*) FROM s1"
    ).fetchone()[0]

    s2_count = con.execute(
        "SELECT COUNT(*) FROM s2"
    ).fetchone()[0]

    print(f"S1 rows: {s1_count:,}")
    print(f"S2 rows: {s2_count:,}")

    # --------------------------------------------------------
    # Candidate count
    # --------------------------------------------------------

    candidate_count = con.execute(f"""
        SELECT COUNT(*)
        FROM read_parquet('{CAND_PATH}')
        WHERE matched_source = 'S2'
    """).fetchone()[0]

    print(f"S2 candidates: {candidate_count:,}")

    # --------------------------------------------------------
    # Stream candidate + source records
    # --------------------------------------------------------

    query = f"""
        SELECT
            c.source1_entity_id,
            c.matched_entity_id,
            s1.country_norm AS s1_country_norm,
            s1.business_name_norm AS s1_business_name_norm,
            s1.business_address_norm AS s1_business_address_norm,
            s1.business_name_clean AS s1_business_name_clean,
            s1.business_address_clean AS s1_business_address_clean,

            s2.country_norm AS match_country_norm,
            s2.business_name_norm AS match_business_name_norm,
            s2.business_address_norm AS match_business_address_norm,
            s2.business_name_clean AS match_business_name_clean,
            s2.business_address_clean AS match_business_address_clean

        FROM read_parquet('{CAND_PATH}') c

        INNER JOIN s1
            ON c.source1_entity_id = s1.entity_id

        INNER JOIN s2
            ON c.matched_entity_id = s2.entity_id

        WHERE c.matched_source = 'S2'
    """

    # --------------------------------------------------------
    # Arrow batch reader
    # --------------------------------------------------------

    reader = con.execute(query).fetch_record_batch(
        rows_per_batch=50_000
    )

    writer = None
    processed = 0

    # --------------------------------------------------------
    # Feature calculation
    # --------------------------------------------------------

    for batch in reader:

        df = batch.read_all().to_pandas()

        feature_rows = []

        for row in df.itertuples(index=False):

            name1 = str(row.s1_business_name_norm)
            name2 = str(row.match_business_name_norm)

            clean_name1 = str(row.s1_business_name_clean)
            clean_name2 = str(row.match_business_name_clean)

            addr1 = str(row.s1_business_address_norm)
            addr2 = str(row.match_business_address_norm)

            clean_addr1 = str(row.s1_business_address_clean)
            clean_addr2 = str(row.match_business_address_clean)

            # Exact number-overlap implementation
            nums1 = set(__import__("re").findall(r"\d+", addr1))
            nums2 = set(__import__("re").findall(r"\d+", addr2))

            if not nums1 and not nums2:
                num_overlap = 1.0
            elif not nums1 or not nums2:
                num_overlap = 0.0
            else:
                num_overlap = len(nums1 & nums2) / len(nums1 | nums2)

            feature_rows.append([
                int(name1 == name2),
                int(clean_name1 == clean_name2),

                ratio(name1, name2) / 100.0,
                token_sort_ratio(name1, name2) / 100.0,
                token_set_ratio(name1, name2) / 100.0,

                int(addr1 == addr2),
                int(clean_addr1 == clean_addr2),

                ratio(addr1, addr2) / 100.0,
                token_sort_ratio(addr1, addr2) / 100.0,
                token_set_ratio(addr1, addr2) / 100.0,

                num_overlap,

                abs(len(name1) - len(name2)),
                abs(len(addr1) - len(addr2)),

                int(row.s1_country_norm == row.match_country_norm),

                2
            ])

        X = pd.DataFrame(
            feature_rows,
            columns=FEATURES
        )

        # ----------------------------------------------------
        # Predict
        # ----------------------------------------------------

        scores = model.predict(X)

        output = pd.DataFrame({
            "source1_entity_id":
                df["source1_entity_id"].astype(str),

            "matched_entity_id":
                df["matched_entity_id"].astype(str),

            "matched_source":
                "S2",

            "score":
                scores.astype(np.float32)
        })

        table = pa.Table.from_pandas(
            output,
            preserve_index=False
        )

        if writer is None:
            writer = pq.ParquetWriter(
                OUTPUT_PATH,
                table.schema,
                compression="zstd"
            )

        writer.write_table(table)

        processed += len(df)

        elapsed = time.time() - start_time
        rate = processed / max(elapsed, 1)

        print(
            f"Processed {processed:,}/{candidate_count:,} "
            f"({processed / candidate_count * 100:.2f}%) "
            f"| {rate:,.0f} rows/sec"
        )

        del df, feature_rows, X, scores, output, table
        gc.collect()

    if writer is not None:
        writer.close()

    con.close()

    elapsed = time.time() - start_time

    print("\n" + "=" * 80)
    print("S2 SCORING COMPLETE")
    print("=" * 80)

    print(f"Rows scored: {processed:,}")
    print(f"Time: {elapsed / 60:.2f} minutes")
    print(f"Output: {OUTPUT_PATH}")
    print(
        f"Size: {os.path.getsize(OUTPUT_PATH) / (1024**2):.2f} MB"
    )
```

```python
# ============================================================
# STEP 3 — SCORE TEST S1 -> SOURCE3 CANDIDATES
# ============================================================

import os
import gc
import time
import re
import duckdb
import numpy as np
import pandas as pd
import pyarrow as pa
import pyarrow.parquet as pq
import lightgbm as lgb

from rapidfuzz.fuzz import (
    ratio,
    token_sort_ratio,
    token_set_ratio
)

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

BASE = "/content/drive/MyDrive/amazon_training"

CAND_PATH = BASE + "/candidates/final_candidate_union.parquet"
S1_PATH = BASE + "/test_prepared/test_s1_prepared.tsv"
S3_PATH = BASE + "/test_prepared/test_s3_prepared.tsv"

OUTPUT_PATH = BASE + "/features/test_scores_source3.parquet"
MODEL_PATH = BASE + "/models/entity_match_lightgbm.txt"

os.makedirs(os.path.dirname(OUTPUT_PATH), exist_ok=True)

# ------------------------------------------------------------
# EXACT TRAINING FEATURE ORDER
# ------------------------------------------------------------

FEATURES = [
    "name_exact",
    "name_clean_exact",
    "name_ratio",
    "name_token_sort",
    "name_token_set",
    "address_exact",
    "address_clean_exact",
    "address_ratio",
    "address_token_sort",
    "address_token_set",
    "number_overlap",
    "name_length_diff",
    "address_length_diff",
    "country_match",
    "source_type"
]

# ------------------------------------------------------------
# RESUME SAFETY
# ------------------------------------------------------------

if os.path.exists(OUTPUT_PATH):

    print("=" * 80)
    print("SOURCE3 SCORES ALREADY EXIST")
    print("=" * 80)

    con = duckdb.connect()

    stats = con.execute(f"""
        SELECT
            COUNT(*) AS rows,
            COUNT(DISTINCT source1_entity_id) AS s1_entities,
            MIN(score) AS min_score,
            MAX(score) AS max_score,
            AVG(score) AS avg_score
        FROM read_parquet('{OUTPUT_PATH}')
    """).fetchone()

    print(f"Rows:        {stats[0]:,}")
    print(f"S1 entities: {stats[1]:,}")
    print(f"Min score:   {stats[2]:.6f}")
    print(f"Max score:   {stats[3]:.6f}")
    print(f"Avg score:   {stats[4]:.6f}")

    con.close()

else:

    # --------------------------------------------------------
    # LOAD MODEL
    # --------------------------------------------------------

    model = lgb.Booster(
        model_file=MODEL_PATH
    )

    print("=" * 80)
    print("STEP 3 — SCORING S1 -> SOURCE3")
    print("=" * 80)

    print("Model loaded.")
    print("Trees:", model.num_trees())

    # --------------------------------------------------------
    # DUCKDB
    # --------------------------------------------------------

    os.makedirs("/content/duckdb_tmp", exist_ok=True)

    con = duckdb.connect()

    con.execute("SET threads=2")
    con.execute("SET memory_limit='5GB'")
    con.execute("SET temp_directory='/content/duckdb_tmp'")
    con.execute("SET preserve_insertion_order=false")

    # --------------------------------------------------------
    # LOAD S1
    # --------------------------------------------------------

    print("\nLoading S1...")

    con.execute(f"""
        CREATE OR REPLACE TEMP TABLE s1 AS
        SELECT
            entity_id,
            country_norm,
            business_name_norm,
            business_address_norm,
            business_name_clean,
            business_address_clean
        FROM read_csv(
            '{S1_PATH}',
            delim='\\t',
            header=true,
            auto_detect=true,
            strict_mode=false
        )
    """)

    # --------------------------------------------------------
    # LOAD SOURCE3
    # --------------------------------------------------------

    print("Loading Source3...")

    con.execute(f"""
        CREATE OR REPLACE TEMP TABLE s3 AS
        SELECT
            entity_id,
            country_norm,
            business_name_norm,
            business_address_norm,
            business_name_clean,
            business_address_clean
        FROM read_csv(
            '{S3_PATH}',
            delim='\\t',
            header=true,
            auto_detect=true,
            strict_mode=false
        )
    """)

    # --------------------------------------------------------
    # COUNTS
    # --------------------------------------------------------

    s1_count = con.execute(
        "SELECT COUNT(*) FROM s1"
    ).fetchone()[0]

    s3_count = con.execute(
        "SELECT COUNT(*) FROM s3"
    ).fetchone()[0]

    candidate_count = con.execute(f"""
        SELECT COUNT(*)
        FROM read_parquet('{CAND_PATH}')
        WHERE matched_source = 'source3'
    """).fetchone()[0]

    print(f"S1 rows:            {s1_count:,}")
    print(f"Source3 rows:       {s3_count:,}")
    print(f"Source3 candidates: {candidate_count:,}")

    if candidate_count == 0:
        raise RuntimeError(
            "Zero source3 candidates found."
        )

    # --------------------------------------------------------
    # JOIN
    # --------------------------------------------------------

    query = f"""
        SELECT
            c.source1_entity_id,
            c.matched_entity_id,

            s1.country_norm AS s1_country_norm,
            s1.business_name_norm AS s1_business_name_norm,
            s1.business_address_norm AS s1_business_address_norm,
            s1.business_name_clean AS s1_business_name_clean,
            s1.business_address_clean AS s1_business_address_clean,

            s3.country_norm AS match_country_norm,
            s3.business_name_norm AS match_business_name_norm,
            s3.business_address_norm AS match_business_address_norm,
            s3.business_name_clean AS match_business_name_clean,
            s3.business_address_clean AS match_business_address_clean

        FROM read_parquet('{CAND_PATH}') c

        INNER JOIN s1
            ON c.source1_entity_id = s1.entity_id

        INNER JOIN s3
            ON c.matched_entity_id = s3.entity_id

        WHERE c.matched_source = 'source3'
    """

    print("\nStarting feature calculation + prediction...")

    reader = con.execute(query).fetch_record_batch(
        rows_per_batch=25_000
    )

    writer = None
    processed = 0
    start_time = time.time()

    # --------------------------------------------------------
    # PROCESS
    # --------------------------------------------------------

    for batch in reader:

        # fetch_record_batch() already returns RecordBatch
        df = batch.to_pandas()

        feature_rows = []

        for row in df.itertuples(index=False):

            name1 = str(row.s1_business_name_norm)
            name2 = str(row.match_business_name_norm)

            clean_name1 = str(row.s1_business_name_clean)
            clean_name2 = str(row.match_business_name_clean)

            addr1 = str(row.s1_business_address_norm)
            addr2 = str(row.match_business_address_norm)

            clean_addr1 = str(row.s1_business_address_clean)
            clean_addr2 = str(row.match_business_address_clean)

            # EXACT ORIGINAL number_overlap
            nums1 = set(
                re.findall(r"\d+", addr1)
            )

            nums2 = set(
                re.findall(r"\d+", addr2)
            )

            if not nums1 and not nums2:
                num_overlap = 1.0

            elif not nums1 or not nums2:
                num_overlap = 0.0

            else:
                num_overlap = (
                    len(nums1 & nums2) /
                    len(nums1 | nums2)
                )

            feature_rows.append([
                int(name1 == name2),
                int(clean_name1 == clean_name2),

                ratio(name1, name2) / 100.0,
                token_sort_ratio(name1, name2) / 100.0,
                token_set_ratio(name1, name2) / 100.0,

                int(addr1 == addr2),
                int(clean_addr1 == clean_addr2),

                ratio(addr1, addr2) / 100.0,
                token_sort_ratio(addr1, addr2) / 100.0,
                token_set_ratio(addr1, addr2) / 100.0,

                num_overlap,

                abs(len(name1) - len(name2)),
                abs(len(addr1) - len(addr2)),

                int(
                    row.s1_country_norm ==
                    row.match_country_norm
                ),

                # source3 = 3
                3
            ])

        X = pd.DataFrame(
            feature_rows,
            columns=FEATURES
        )

        scores = model.predict(X)

        output = pd.DataFrame({
            "source1_entity_id":
                df["source1_entity_id"].astype(str),

            "matched_entity_id":
                df["matched_entity_id"].astype(str),

            "matched_source":
                "source3",

            "score":
                scores.astype(np.float32)
        })

        table = pa.Table.from_pandas(
            output,
            preserve_index=False
        )

        if writer is None:
            writer = pq.ParquetWriter(
                OUTPUT_PATH,
                table.schema,
                compression="zstd"
            )

        writer.write_table(table)

        processed += len(df)

        elapsed = time.time() - start_time
        rate = processed / max(elapsed, 1)

        print(
            f"Processed {processed:,}/{candidate_count:,} "
            f"({processed / candidate_count * 100:.2f}%) "
            f"| {rate:,.0f} rows/sec"
        )

        del df
        del feature_rows
        del X
        del scores
        del output
        del table

        gc.collect()

    # --------------------------------------------------------
    # CLOSE
    # --------------------------------------------------------

    if writer is not None:
        writer.close()

    con.close()

    elapsed = time.time() - start_time

    print("\n" + "=" * 80)
    print("SOURCE3 SCORING COMPLETE")
    print("=" * 80)

    print(f"Rows scored: {processed:,}")
    print(f"Time: {elapsed / 60:.2f} minutes")
    print(f"Output: {OUTPUT_PATH}")

    if os.path.exists(OUTPUT_PATH):

        print(
            f"Size: "
            f"{os.path.getsize(OUTPUT_PATH)/(1024**2):.2f} MB"
        )

    else:
        raise RuntimeError(
            "Output parquet was not created."
        )
```

```python
import pandas as pd
import duckdb
import numpy as np

BASE = "/content/drive/MyDrive/amazon_training"
SCORE2 = f"{BASE}/features/test_scores_source2.parquet"
SCORE3 = f"{BASE}/features/test_scores_source3.parquet"
CAND = f"{BASE}/candidates/final_candidate_union.parquet"

print("=" * 80)
print("STEP 7 — RETRIEVE FEATURE EVIDENCE")
print("=" * 80)

# -------------------------------------------------------------------------
# 1. Load scores
# -------------------------------------------------------------------------

s2 = pd.read_parquet(SCORE2)
s3 = pd.read_parquet(SCORE3)

scores = pd.concat([s2, s3], ignore_index=True)

# -------------------------------------------------------------------------
# 2. Select a SMALL diagnostic sample
# -------------------------------------------------------------------------
# We intentionally do not process all 12M rows again.

scores = scores.sort_values(
    ["source1_entity_id", "score"],
    ascending=[True, False]
)

scores["rank"] = scores.groupby("source1_entity_id").cumcount() + 1

# Top-1 candidates
top1 = scores[scores["rank"] == 1].copy()

# Pick representative cases:
#   A) very high confidence
#   B) ambiguous top candidates
#   C) low-score top candidates

high = top1.nlargest(20, "score")

low = top1.nsmallest(20, "score")

# Candidates whose top score is high but whose second candidate is very close
top2 = scores[scores["rank"] == 2][
    ["source1_entity_id", "score"]
].rename(columns={"score": "score_2"})

ambiguous = top1.merge(
    top2,
    on="source1_entity_id",
    how="left"
)

ambiguous["score_2"] = ambiguous["score_2"].fillna(0)
ambiguous["gap"] = ambiguous["score"] - ambiguous["score_2"]

ambiguous = ambiguous.nsmallest(20, "gap")

sample_ids = pd.concat([
    high,
    ambiguous,
    low
], ignore_index=True)[
    ["source1_entity_id", "matched_entity_id", "matched_source", "score"]
].drop_duplicates()

print(f"\nDiagnostic candidates selected: {len(sample_ids):,}")

# -------------------------------------------------------------------------
# 3. Open the existing TEST source files through DuckDB
# -------------------------------------------------------------------------

con = duckdb.connect()

s1_path = f"{BASE}/test_prepared/test_s1_prepared.tsv"
s2_path = f"{BASE}/test_prepared/test_s2_prepared.tsv"
s3_path = f"{BASE}/test_prepared/test_s3_prepared.tsv"

# -------------------------------------------------------------------------
# 4. Retrieve S1 records
# -------------------------------------------------------------------------

s1_ids = sample_ids["source1_entity_id"].tolist()

# Create temporary table rather than constructing a giant SQL IN clause.
con.register("sample_s1", pd.DataFrame({
    "entity_id": s1_ids
}))

s1 = con.execute(f"""
    SELECT
        t.entity_id,
        t.country_norm,
        t.business_name_norm,
        t.business_address_norm,
        t.business_name_clean,
        t.business_address_clean,
        t.address_numbers
    FROM read_csv_auto(
        '{s1_path}',
        delim='\\t',
        header=true
    ) t
    INNER JOIN sample_s1 x
        ON t.entity_id = x.entity_id
""").fetchdf()

# -------------------------------------------------------------------------
# 5. Retrieve candidate records
# -------------------------------------------------------------------------

s2_ids = sample_ids.loc[
    sample_ids["matched_source"] == "source2",
    "matched_entity_id"
].tolist()

s3_ids = sample_ids.loc[
    sample_ids["matched_source"] == "source3",
    "matched_entity_id"
].tolist()

if s2_ids:
    con.register("sample_s2", pd.DataFrame({
        "entity_id": s2_ids
    }))

    cand_s2 = con.execute(f"""
        SELECT
            t.entity_id,
            t.country_norm,
            t.business_name_norm,
            t.business_address_norm,
            t.business_name_clean,
            t.business_address_clean,
            t.address_numbers
        FROM read_csv_auto(
            '{s2_path}',
            delim='\\t',
            header=true
        ) t
        INNER JOIN sample_s2 x
            ON t.entity_id = x.entity_id
    """).fetchdf()
else:
    cand_s2 = pd.DataFrame()

if s3_ids:
    con.register("sample_s3", pd.DataFrame({
        "entity_id": s3_ids
    }))

    cand_s3 = con.execute(f"""
        SELECT
            t.entity_id,
            t.country_norm,
            t.business_name_norm,
            t.business_address_norm,
            t.business_name_clean,
            t.business_address_clean,
            t.address_numbers
        FROM read_csv_auto(
            '{s3_path}',
            delim='\\t',
            header=true
        ) t
        INNER JOIN sample_s3 x
            ON t.entity_id = x.entity_id
    """).fetchdf()
else:
    cand_s3 = pd.DataFrame()

candidates = pd.concat(
    [cand_s2.assign(matched_source="source2"),
     cand_s3.assign(matched_source="source3")],
    ignore_index=True
)

# -------------------------------------------------------------------------
# 6. Merge S1 + candidate data
# -------------------------------------------------------------------------

result = sample_ids.merge(
    s1.add_prefix("s1_"),
    left_on="source1_entity_id",
    right_on="s1_entity_id",
    how="left"
)

result = result.merge(
    candidates.add_prefix("cand_"),
    left_on=["matched_entity_id", "matched_source"],
    right_on=["cand_entity_id", "cand_matched_source"],
    how="left"
)

# -------------------------------------------------------------------------
# 7. Calculate the SAME matching features used for training
# -------------------------------------------------------------------------

from rapidfuzz.fuzz import ratio, token_sort_ratio, token_set_ratio
import re

def number_set(text):
    return set(re.findall(r"\d+", str(text)))

def number_overlap(a, b):
    a_nums = number_set(a)
    b_nums = number_set(b)

    if not a_nums and not b_nums:
        return 1.0
    if not a_nums or not b_nums:
        return 0.0

    return len(a_nums & b_nums) / len(a_nums | b_nums)

def safe_ratio(a, b, func):
    return func(str(a), str(b)) / 100.0

feature_rows = []

for _, r in result.iterrows():

    s1_name = str(r["s1_business_name_norm"])
    c_name = str(r["cand_business_name_norm"])

    s1_clean_name = str(r["s1_business_name_clean"])
    c_clean_name = str(r["cand_business_name_clean"])

    s1_addr = str(r["s1_business_address_norm"])
    c_addr = str(r["cand_business_address_norm"])

    s1_clean_addr = str(r["s1_business_address_clean"])
    c_clean_addr = str(r["cand_business_address_clean"])

    feature_rows.append({
        "name_exact":
            int(s1_name == c_name),

        "name_clean_exact":
            int(s1_clean_name == c_clean_name),

        "name_ratio":
            safe_ratio(s1_name, c_name, ratio),

        "name_token_sort":
            safe_ratio(s1_name, c_name, token_sort_ratio),

        "name_token_set":
            safe_ratio(s1_name, c_name, token_set_ratio),

        "address_exact":
            int(s1_addr == c_addr),

        "address_clean_exact":
            int(s1_clean_addr == c_clean_addr),

        "address_ratio":
            safe_ratio(s1_addr, c_addr, ratio),

        "address_token_sort":
            safe_ratio(s1_addr, c_addr, token_sort_ratio),

        "address_token_set":
            safe_ratio(s1_addr, c_addr, token_set_ratio),

        "number_overlap":
            number_overlap(
                r["s1_business_address_norm"],
                r["cand_business_address_norm"]
            ),

        "name_length_diff":
            abs(len(s1_name) - len(c_name)),

        "address_length_diff":
            abs(len(s1_addr) - len(c_addr)),

        "country_match":
            int(r["s1_country_norm"] == r["cand_country_norm"]),

        "source_type":
            2 if r["matched_source"] == "source2" else 3
    })

features = pd.DataFrame(feature_rows)

final = pd.concat(
    [
        result[
            [
                "source1_entity_id",
                "matched_entity_id",
                "matched_source",
                "score"
            ]
        ].reset_index(drop=True),
        features
    ],
    axis=1
)

print("\nFEATURE EVIDENCE")
print("-" * 80)

print(
    final.to_string(index=False)
)

print("\n" + "=" * 80)
print("STEP 7 COMPLETE")
print("=" * 80)
```

```python
import duckdb
import os

BASE = "/content/drive/MyDrive/amazon_training"

SCORE2 = f"{BASE}/features/test_scores_source2.parquet"
SCORE3 = f"{BASE}/features/test_scores_source3.parquet"
PROV = f"{BASE}/candidates/final_candidate_provenance.parquet"

OUT_DIR = f"{BASE}/submission"
os.makedirs(OUT_DIR, exist_ok=True)

MATCHING = f"{OUT_DIR}/matching_results.tsv"
CANDIDATES = f"{OUT_DIR}/candidate_pairs.tsv"

con = duckdb.connect()

print("=" * 80)
print("FINAL SUBMISSION GENERATION")
print("=" * 80)

# ------------------------------------------------------------------
# 1. Combine existing scores + provenance
# ------------------------------------------------------------------

con.execute(f"""
CREATE OR REPLACE TEMP TABLE scored AS
SELECT
    s.source1_entity_id,
    s.matched_entity_id,
    s.matched_source,
    s.score,
    p.from_exact_name,
    p.from_exact_clean_name,
    p.from_exact_address
FROM (
    SELECT * FROM read_parquet('{SCORE2}')
    UNION ALL
    SELECT * FROM read_parquet('{SCORE3}')
) s
JOIN read_parquet('{PROV}') p
ON  s.source1_entity_id = p.source1_entity_id
AND s.matched_entity_id = p.matched_entity_id
AND s.matched_source = p.matched_source
""")

# ------------------------------------------------------------------
# 2. Rank candidates within every Source1
# ------------------------------------------------------------------

con.execute("""
CREATE OR REPLACE TEMP TABLE ranked AS
SELECT *,
       ROW_NUMBER() OVER (
           PARTITION BY source1_entity_id
           ORDER BY score DESC
       ) AS rn
FROM scored
""")

# ------------------------------------------------------------------
# 3. Conservative final selection
#
#    A = exact address
#    B = exact normalized + clean name
#    C = strongest model candidate above 0.82
# ------------------------------------------------------------------

con.execute("""
CREATE OR REPLACE TEMP TABLE selected AS
SELECT *
FROM ranked
WHERE
      from_exact_address = 1
   OR (
        from_exact_name = 1
        AND from_exact_clean_name = 1
      )
   OR (
        rn = 1
        AND score >= 0.82
      )
""")

selected_count = con.execute(
    "SELECT COUNT(*) FROM selected"
).fetchone()[0]

selected_s1 = con.execute(
    "SELECT COUNT(DISTINCT source1_entity_id) FROM selected"
).fetchone()[0]

print(f"\nSelected matches : {selected_count:,}")
print(f"S1 with matches  : {selected_s1:,}")

# ------------------------------------------------------------------
# 4. Candidate pairs submission
# ------------------------------------------------------------------

con.execute(f"""
COPY (
    SELECT
        source1_entity_id,
        matched_entity_id,
        matched_source
    FROM ranked
    ORDER BY source1_entity_id, rn
)
TO '{CANDIDATES}'
(FORMAT CSV, DELIMITER '\\t', HEADER TRUE)
""")

print(f"\nCandidate pairs written:")
print(CANDIDATES)

# ------------------------------------------------------------------
# 5. Every Source1 must appear in matching_results
#
#    Read test Source1 IDs directly from prepared file.
# ------------------------------------------------------------------

S1 = f"{BASE}/test_prepared/test_s1_prepared.tsv"

con.execute(f"""
CREATE OR REPLACE TEMP TABLE all_s1 AS
SELECT entity_id AS source1_entity_id
FROM read_csv_auto(
    '{S1}',
    delim='\\t',
    header=true
)
""")

# ------------------------------------------------------------------
# 6. Aggregate selected matches
# ------------------------------------------------------------------

con.execute("""
CREATE OR REPLACE TEMP TABLE aggregated AS
SELECT
    source1_entity_id,
    STRING_AGG(
        matched_entity_id,
        ',' ORDER BY score DESC
    ) AS matched_entity_ids
FROM selected
GROUP BY source1_entity_id
""")

# ------------------------------------------------------------------
# 7. Final matching_results.tsv
# ------------------------------------------------------------------

con.execute(f"""
COPY (
    SELECT
        a.source1_entity_id,
        COALESCE(g.matched_entity_ids, '') AS matched_entity_ids
    FROM all_s1 a
    LEFT JOIN aggregated g
        ON a.source1_entity_id = g.source1_entity_id
    ORDER BY a.source1_entity_id
)
TO '{MATCHING}'
(FORMAT CSV, DELIMITER '\\t', HEADER TRUE)
""")

# ------------------------------------------------------------------
# 8. Final checks
# ------------------------------------------------------------------

total_s1 = con.execute(
    "SELECT COUNT(*) FROM all_s1"
).fetchone()[0]

result_s1 = con.execute(f"""
SELECT COUNT(*)
FROM read_csv_auto(
    '{MATCHING}',
    delim='\\t',
    header=true
)
""").fetchone()[0]

print("\nFINAL CHECK")
print("-" * 80)
print(f"Expected Source1 entities : {total_s1:,}")
print(f"Output Source1 entities   : {result_s1:,}")
print(f"Selected matches          : {selected_count:,}")

print("\nFILES:")
print(MATCHING)
print(CANDIDATES)

print("\n" + "=" * 80)
print("SUBMISSION FILES GENERATED")
print("=" * 80)
```

```python
import pandas as pd
import numpy as np

BASE = "/content/drive/MyDrive/amazon_training"

TRAIN = f"{BASE}/features/train_features.tsv"
VALID = f"{BASE}/features/valid_features.tsv"

FEATURES = [
    "name_exact",
    "name_clean_exact",
    "name_ratio",
    "name_token_sort",
    "name_token_set",
    "address_exact",
    "address_clean_exact",
    "address_ratio",
    "address_token_sort",
    "address_token_set",
    "number_overlap",
    "name_length_diff",
    "address_length_diff",
    "country_match",
    "source_type"
]

print("=" * 80)
print("SUPERVISED SELECTION CHECK")
print("=" * 80)

train = pd.read_csv(TRAIN, sep="\t")
valid = pd.read_csv(VALID, sep="\t")

print("Train:", train.shape)
print("Valid:", valid.shape)

# Train LightGBM model
import lightgbm as lgb

model = lgb.Booster(
    model_file=f"{BASE}/models/entity_match_lightgbm.txt"
)

Xv = valid[FEATURES]
yv = valid["label"].values

valid["score"] = model.predict(Xv)

# Rank candidates inside each S1
valid = valid.sort_values(
    ["source1_entity_id", "score"],
    ascending=[True, False]
)

valid["rank"] = (
    valid.groupby("source1_entity_id")
    .cumcount() + 1
)

valid["best_score"] = (
    valid.groupby("source1_entity_id")["score"]
    .transform("max")
)

valid["gap"] = valid["best_score"] - valid["score"]

# ------------------------------------------------------------------
# Macro F0.5
# ------------------------------------------------------------------

def macro_f05(df, mask):
    selected = df[mask]

    grouped = selected.groupby("source1_entity_id")

    scores = []

    for s1, g in df.groupby("source1_entity_id"):

        true_ids = set(
            g.loc[g["label"] == 1, "matched_entity_id"]
        )

        pred_ids = set(
            selected.loc[
                selected["source1_entity_id"] == s1,
                "matched_entity_id"
            ]
        )

        tp = len(true_ids & pred_ids)
        fp = len(pred_ids - true_ids)
        fn = len(true_ids - pred_ids)

        if tp == 0:
            f05 = 0.0 if (fp > 0 or fn > 0) else 1.0
        else:
            precision = tp / (tp + fp)
            recall = tp / (tp + fn)

            f05 = (
                1.25 * precision * recall /
                (0.25 * precision + recall)
            )

        scores.append(f05)

    return np.mean(scores)


rules = {
    "top1": valid["rank"] == 1,

    "top1_score_082":
        (valid["rank"] == 1) &
        (valid["score"] >= 0.82),

    "top1_score_090":
        (valid["rank"] == 1) &
        (valid["score"] >= 0.90),

    "top1_score_095":
        (valid["rank"] == 1) &
        (valid["score"] >= 0.95),

    "gap_0001":
        (valid["score"] >= 0.82) &
        (valid["gap"] <= 0.0001),

    "gap_001":
        (valid["score"] >= 0.82) &
        (valid["gap"] <= 0.001),

    "exact_address":
        valid["address_exact"] == 1,

    "exact_name_and_address":
        (valid["name_exact"] == 1) &
        (valid["address_exact"] == 1),

    "exact_clean_name_and_address":
        (valid["name_clean_exact"] == 1) &
        (valid["address_exact"] == 1),
}

print("\nRULE PERFORMANCE")
print("-" * 80)

for name, mask in rules.items():

    selected = valid[mask]

    # Pair precision
    precision = (
        selected["label"].mean()
        if len(selected)
        else 0
    )

    print(
        f"{name:35s} "
        f"pairs={len(selected):6,} "
        f"precision={precision:.4f}"
    )

print("\nNOTE:")
print("Pairwise validation is only a diagnostic here;")
print("the final challenge metric is per-Source1 Macro F0.5.")

print("\n" + "=" * 80)
print("SELECTION CHECK COMPLETE")
print("=" * 80)
```
